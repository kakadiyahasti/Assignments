Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for 12 friends
friends = [
    'Aarav',
    'Priya',
    'Rohan',
    'Sneha',
    'Vikram',
    'Ananya',
    'Kabir',
    'Neha',
    'Amit',
    'Pooja',
    'Rahul',
    'Simran',
]
orders = np.array([5, 12, 8, 15, 3, 20, 10, 7, 14, 6, 18, 9], dtype=float)
amount_spent = np.array(
    [1200, 3500, 2100, 4800, 800, 6200, 2900, 1800, 4100, 1500, 5500, 2600],
    dtype=float,
)

# Calculate best-fit line (degree 1 polynomial: y = mx + c)
slope, intercept = np.polyfit(orders, amount_spent, 1)
trend_line = slope * orders + intercept

# Create scatter plot
plt.figure(figsize=(10, 7))
plt.scatter(
    orders,
    amount_spent,
    color='#ff6b6b',
    s=100,
    edgecolor='black',
    alpha=0.85,
    label='Friends',
)

# Plot the trend line
plt.plot(
    orders,
    trend_line,
    color='#1f77b4',
    linewidth=2.5,
    linestyle='--',
    label=f'Trend Line: y = {slope:.2f}x + {intercept:.2f}',
)

# Annotate each point with the friend's name
for i, txt in enumerate(friends):
  plt.annotate(
      txt,
      (orders[i], amount_spent[i]),
      xytext=(5, 5),
      textcoords='offset points',
      fontsize=10,
      fontweight='medium',
      color='#333333',
  )

# Add title, axis labels, and legend
plt.title(
    'Zomato Usage: Orders vs. Amount Spent with Trend Line',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('Number of Orders Placed', fontsize=12)
plt.ylabel('Amount Spent on Zomato (₹)', fontsize=12)
plt.legend(fontsize=11, loc='upper left')

# Add grid for readability
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for 15 users
youtube_hours = np.array(
    [1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 1.8, 2.2, 3.2, 3.8, 4.2, 4.8, 5.2],
    dtype=float,
)
spotify_hours = np.array(
    [
        1.45,
        1.53,
        2.32,
        3.16,
        2.69,
        3.08,
        4.38,
        4.38,
        1.21,
        2.03,
        2.33,
        2.81,
        3.48,
        2.88,
        3.29,
    ],
    dtype=float,
)

# Calculate best-fit line (degree 1 polynomial: y = mx + c)
slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)
trend_line = slope * youtube_hours + intercept

# Create scatter plot
plt.figure(figsize=(10, 7))
plt.scatter(
    youtube_hours,
    spotify_hours,
    color='#ff6b6b',
    s=100,
    edgecolor='black',
    alpha=0.85,
    label='Users',
)

# Plot the regression trend line with the equation
equation_str = f'Trend Line: y = {slope:.2f}x + {intercept:.2f}'
plt.plot(
    youtube_hours,
    trend_line,
    color='#1f77b4',
    linewidth=2.5,
    linestyle='--',
    label=equation_str,
)

# Add title, axis labels, and legend
plt.title(
    'Daily Media Consumption: YouTube Watch Hours vs. Spotify Listening Hours',
    fontsize=13,
    fontweight='bold',
)
plt.xlabel('YouTube Watch Hours per Day', fontsize=12)
plt.ylabel('Spotify Listening Hours per Day', fontsize=12)
plt.legend(fontsize=11, loc='upper left')

# Add grid for readability
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()

# Print the exact regression equation
print(f'Fitted Regression Equation: y = {slope:.4f}x + {intercept:.4f}')
```

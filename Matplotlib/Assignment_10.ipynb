```python
import matplotlib.pyplot as plt 
```

Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
dau = [42.0, 43.5, 45.0, 46.2, 48.0, 47.5, 49.0, 51.2, 53.0, 55.5, 54.0, 58.5]

plt.plot(months, dau, marker='o', color='#1DB954', linewidth=2.5, markersize=6, label='Monthly Average DAU')

max_dau = max(dau)
max_index = dau.index(max_dau)
max_month = months[max_index]

plt.annotate(
    f'Peak: {max_dau}M ({max_month})',
    xy=(max_index, max_dau),                       # Point to annotate
    xytext=(max_index - 2.2, max_dau - 6),         # Text position
    arrowprops=dict(facecolor='#333333', arrowstyle='->', lw=1.5),
    fontsize=10,
    fontweight='bold',
    bbox=dict(boxstyle='round,pad=0.4', facecolor='#E8F5E9', edgecolor='#1DB954', alpha=0.9)
)

plt.show()
```

Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>

```python
# 12 months of the year
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# Simulated monthly food orders (in millions)
orders = [45, 38, 32, 44, 52, 58, 65, 60, 55, 50, 48, 62]

# Set the figure size
plt.figure(figsize=(11, 5))

# Plot the bar chart using Zomato crimson red
bars = plt.bar(months, orders, color='#CB202D', width=0.6)

# Find the month with the lowest orders
min_orders = min(orders)
min_index = orders.index(min_orders)
min_month = months[min_index]

# Add an arrow annotation using plt.annotate() and arrowprops
plt.annotate(
    f'Lowest: {min_orders}M ({min_month})',
    xy=(min_index, min_orders),                    # Coordinates of the lowest bar tip
    xytext=(min_index + 1.2, min_orders + 15),       # Position of the text box
    arrowprops=dict(facecolor='#333333', arrowstyle='->', lw=1.5),
    fontsize=10,
    fontweight='bold',
    bbox=dict(boxstyle='round,pad=0.4', facecolor='#FFEBEE', edgecolor='#CB202D', alpha=0.9)
)

# Add titles and labels
plt.title('Monthly Food Delivery Orders Trend (Zomato)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Month', fontsize=11)
plt.ylabel('Total Orders (in Millions)', fontsize=11)

# Add a horizontal grid for readability
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Adjust layout to prevent label clipping
plt.tight_layout()

# Display the chart
plt.show()
```

Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.

```python
# Payment methods and their corresponding usage percentages
payment_methods = ['UPI', 'Debit/Credit Card', 'Mobile Wallet', 'Cash on Delivery']
usage_shares = [55.0, 25.0, 12.0, 8.0]

# Flipkart-inspired color palette
colors = ['#2874F0', '#FF9F00', '#48C9B0', '#E74C3C']

# Set the figure size (slightly square for a balanced pie chart)
plt.figure(figsize=(4,4))

# Plot the pie chart
wedges, texts, autotexts = plt.pie(
    usage_shares, 
    labels=payment_methods, 
    colors=colors, 
    autopct='%1.1f%%', 
    startangle=140, 
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 10, 'fontweight': 'bold'}
)

# Customize percentage text color for better contrast
for autotext in autotexts:
    autotext.set_color('white')

# Add a title
plt.title('Payment Method Distribution (Flipkart E-Commerce)', fontsize=14, fontweight='bold', pad=20)

# Add a legend positioned at the lower right corner
plt.legend(wedges, payment_methods, loc='lower right', fontsize=10, frameon=True)

# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')

# Adjust layout to prevent clipping
plt.tight_layout()

# Display the chart
plt.show()
```

Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>

```python
# Sample movie data
movies = ['Blockbuster X', 'Drama Y', 'Comedy Z', 'Action Alpha', 'Thriller Beta', 'Romance Gamma', 'Sci-Fi Delta']
reviews = [45, 18, 32, 85, 22, 60, 50]  # in thousands
ratings = [8.5, 7.2, 8.1, 9.4, 7.8, 8.9, 8.6]

# Set figure size
plt.figure(figsize=(10, 6))

# Plot the scatter plot using BookMyShow red
plt.scatter(
    reviews, 
    ratings, 
    color='#C40030', 
    s=120, 
    alpha=0.85, 
    edgecolor='black', 
    linewidth=1, 
    label='Movies'
)

# Find the movie with the highest rating
max_rating = max(ratings)
max_index = ratings.index(max_rating)
max_movie = movies[max_index]
max_reviews = reviews[max_index]

# Use plt.annotate() with the 'bbox' parameter to highlight the top-rated movie
plt.annotate(
    f'Top Rated:\n{max_movie} ({max_rating}★)',
    xy=(max_reviews, max_rating),                  # Coordinates of the point
    xytext=(max_reviews - 22, max_rating - 0.7),   # Text box position
    arrowprops=dict(facecolor='#333333', arrowstyle='->', lw=1.5),
    fontsize=10,
    fontweight='bold',
    bbox=dict(
        boxstyle='round,pad=0.5', 
        facecolor='#FFF0F2', 
        edgecolor='#C40030', 
        alpha=0.9, 
        linewidth=1.5
    )
)

# Add titles and labels
plt.title('Movie Ratings vs. Number of Reviews (BookMyShow)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Reviews (in Thousands)', fontsize=11)
plt.ylabel('Rating (Out of 10)', fontsize=11)

# Add grid and legend
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='lower right', fontsize=11, frameon=True)

# Adjust layout to prevent clipping
plt.tight_layout()

# Display the chart
plt.show()
```

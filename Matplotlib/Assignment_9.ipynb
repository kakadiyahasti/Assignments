Create a stacked bar chart using Matplotlib that shows the number of food orders from Zomato and Swiggy for each day of the week. Use different colors for each platform and add a legend.

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
zomato_orders = [2.1, 2.3, 2.4, 2.5, 3.2, 3.8, 3.6]
swiggy_orders = [1.9, 2.1, 2.2, 2.3, 3.0, 3.6, 3.4]

plt.bar(days,zomato_orders, color='#5E4A9C', label='zomato')
plt.bar(days,swiggy_orders, bottom=zomato_orders,color='#6BAACE', label='swiggy' )
plt.legend()
plt.show()
```

Plot an area chart that visualizes the daily active users on Spotify and YouTube over a 7-day period using Matplotlib's fill_between(). Make sure each platform's area is filled with a different color and add labels for clarity.

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
youtube_dau = [120.0, 122.5, 124.0, 126.5, 132.0, 140.0, 138.0]
spotify_dau = [42.5, 43.1, 44.0, 43.8, 46.2, 49.5, 48.0]

plt.plot(days,youtube_dau,color='green',label='youtube')
plt.fill_between(days,youtube_dau,color='green',alpha=0.4)

plt.plot(days,spotify_dau,color='blue',label='spotify')
plt.fill_between(days,spotify_dau,color='blue',alpha=0.3)

plt.legend()
plt.show()
```

Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard deviation for each day's delivery time using Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr parameter to show error bars for each data point.</em>

```python
days = [f'Day {i}' for i in range(1, 11)]
avg_delivery_time = [45, 42, 38, 40, 35, 33, 36, 32, 30, 28]
std_dev = [5.2, 4.8, 3.5, 4.0, 3.1, 2.5, 3.0, 2.2, 2.0, 1.8]

plt.figure(figsize=(10,5))

plt.errorbar(
    days, 
    avg_delivery_time, 
    yerr=std_dev, 
    fmt='-o',             
    color='#2874F0',      
    ecolor='orange',        
    elinewidth=1.5,       
    capsize=5,            
    linewidth=2,          
    markersize=6          
)

plt.show()
```

Plot a time series line chart showing your daily Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to display dates in 'DD MMM' format (e.g., '12 Jun').

```python
import matplotlib.dates as mdates
from datetime import datetime, timedelta

end_date = datetime(2026, 10, 22)
dates = [end_date - timedelta(days=i) for i in range(13, -1, -1)]

wallet_balance = [
    3450, 4200, 3900, 5100, 4800, 
    6200, 5800, 7100, 6500, 8200, 
    7900, 9400, 8800, 10500
]
plt.plot(dates, wallet_balance, marker='o', color='#002E6E', linewidth=2.5, markersize=6, label='Wallet Balance')

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%d %b'))
plt.gca().xaxis.set_major_locator(mdates.DayLocator(interval=1))

# Rotate x-axis labels slightly for clean readability
plt.gcf().autofmt_xdate(rotation=30)

plt.show()
```

Refactor your area chart code to add gridlines and custom axis titles, making the chart look more like the analytics dashboards you see in apps like Instagram Insights.<br><br><em><strong>Constraint:</strong> Use ax.set_title(), ax.set_xlabel(), ax.set_ylabel(), and ax.grid(True) in your code.</em>

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
youtube_dau = [120.0, 122.5, 124.0, 126.5, 132.0, 140.0, 138.0]
spotify_dau = [42.5, 43.1, 44.0, 43.8, 46.2, 49.5, 48.0]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(days, youtube_dau, label='YouTube', color='#FF0000', linewidth=2)
ax.fill_between(days, youtube_dau, color='#FF0000', alpha=0.2)

ax.plot(days, spotify_dau, label='Spotify', color='#1DB954', linewidth=2)
ax.fill_between(days, spotify_dau, color='#1DB954', alpha=0.3)

# Configure titles and axes using explicit ax methods
ax.set_title('Weekly Audience Reach & Engagement (Dashboard View)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Day of the Week', fontsize=11)
ax.set_ylabel('Active Users (in Millions)', fontsize=11)

# Enable custom gridlines
ax.grid(True, linestyle='--', alpha=0.5)

# Add legend
ax.legend(loc='upper left', fontsize=11, frameon=True)

# Adjust layout to prevent label clipping
plt.tight_layout()

plt.show()
```

Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.

```python
weekdays = ['monday','tuesday','wednesday','thursday','friday','saturday','sunday']
spotify = [2.5,1,4,3.5,1.5,2,3.5]
youtube = [2,4,3,4.5,2.5,1,3.5]

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.bar(weekdays, spotify, color='yellow')
plt.title('spotify listening hours')
plt.xlabel("weekdays")
plt.xticks(rotation =45, ha='right')
plt.ylabel('listening hours')

plt.subplot(1,2,2)
plt.bar(weekdays, youtube, color='orange')
plt.title('youtube watching hours')
plt.xlabel("weekdays")
plt.xticks(rotation =45, ha='right')
plt.ylabel('watching hours')

plt.show()
```

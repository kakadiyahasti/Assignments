```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. Set the line color to green and linewidth to 2.

```python
steps = [345,456,543,665,765,876,230]
days = ['day 1','day 2','day 3','day 4','day 5','day 6','day 7']

plt.figure(figsize=(6,4))
plt.plot(days, steps, color='green', linewidth=2, marker='o')
plt.title('daily steps count')
plt.xlabel("steps")
plt.ylabel("days")
plt.tight_layout()
plt.show()
```

Plot a line chart of your favorite music app's monthly active users for 6 months using plt.plot(). Add circle markers at each data point and set the marker color to red.<br><br><em><strong>Hint:</strong> Use the marker and markerfacecolor parameters in plt.plot().</em>

```python
data = np.random.randint(345674,678543,(6))
users = data.tolist()
months = ['jan' , 'feb', 'march', 'apr' , 'may', 'june']

plt.figure(figsize=(10,4))
plt.plot(months, users,color = 'brown', marker='o', markerfacecolor = 'red')
plt.title('monthly active users for 6 months')
plt.xlabel("months")
plt.ylabel("user")
plt.tight_layout()
plt.show()
```

Create a line chart comparing the number of orders you placed on Zomato and Swiggy over the last 5 months. Use different line styles and colors for each app, and add a legend to distinguish them.

```python
zomato = [12,13,14,10,8]
swiggy = [7,15,12,9,7]
months = ['jan' , 'feb', 'march', 'apr' , 'may']

plt.figure(figsize=(10,4))
plt.plot(months, zomato, color='red', marker='o', label='Zomato')
plt.plot(months, swiggy, color='green',linestyle='-.', label='Swiggy')
plt.title('zomato and swiggy order over last 5 months')
plt.xlabel("months")
plt.ylabel("orders")
plt.legend(['zomato','swiggy'])
plt.tight_layout()
plt.show()
```

Add a main title 'Monthly Data Usage Trend' and label the X-axis as 'Month' and Y-axis as 'Data Used (GB)' to a line chart showing your mobile data usage for the past 6 months.<br><br><em><strong>Hint:</strong> Use plt.title(), plt.xlabel(), and plt.ylabel().</em>

```python
months = ['jan' , 'feb', 'march', 'apr' , 'may', 'june']
data_usage = [59,78,60,84,90,60]

plt.figure(figsize=(10,4))
plt.plot(months, data_usage,color = 'purple', marker='o', markerfacecolor = 'purple')
plt.title('monthly data usage for 6 months')
plt.xlabel("months")
plt.ylabel("data used")
plt.tight_layout()
plt.show()
```

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

Use plt.pie() to visualize the percentage of time you spend on 5 different apps in a day (e.g., Instagram, YouTube, WhatsApp, Zomato, Spotify) and display the percentage values on each segment using the autopct parameter.

```python
apps = ['YouTube', 'Instagram', 'WhatsApp', 'Spotify', 'Zomato']
hours = [3.0, 2.5, 1.5, 1.0, 0.5]



plt.pie(
    hours,
    labels=apps,
    autopct='%1.1f%%',
    startangle=140,
    colors=['#66b3ff', '#ff9999', '#99ff99', '#ffcc99', '#c2c2f0'],
)

plt.tight_layout()
plt.show()
```

Create a pie chart showing the distribution of your monthly online spending across 4 categories: Food Delivery, Shopping, Entertainment, and UPI Payments, and use the explode parameter to highlight the category with the highest spending.<br><br><em><strong>Hint:</strong> Set a higher explode value for the segment you want to pull out from the pie.

```python
categories = ['Shopping', 'Food Delivery', 'UPI Payments', 'Entertainment']
spending = [4500, 3000, 2500, 1500]

explode = [0.1,0,0,0]

plt.pie(
    spending,
    labels=categories,
    explode=explode,
    startangle=140,
    autopct="%1.1f%%",
    colors=['#66b3ff', '#ff9999', '#99ff99', '#ffcc99'],
    shadow=True,
)

plt.tight_layout()
plt.show()
```

Transform your pie chart into a donut chart to show the same spending data by adding a white circle at the center using plt.Circle and ax.add_artist.

```python
categories = ['Shopping', 'Food Delivery', 'UPI Payments', 'Entertainment']
spending = [4500, 3000, 2500, 1500]

explode = [0.1,0,0,0]
fig, ax = plt.subplots(figsize=(4,4))
plt.pie(
    spending,
    labels=categories,
    explode=explode,
    startangle=140,
    autopct="%1.1f%%",
    colors=['#66b3ff', '#ff9999', '#99ff99', '#ffcc99'],
    shadow=True,
)
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
ax.add_artist(centre_circle)

plt.tight_layout()
plt.show()
```

Suppose you want to compare the market share of 4 IPL cricket teams based on the number of Instagram followers. Build a donut chart where each segment shows the team name and follower percentage, and the chart title clearly states 'IPL Teams Instagram Followers Share'.<br><br><em><strong>Constraint:</strong> Use both labels and autopct to display information on the chart.</em>

```python
teams = [
    'Chennai Super Kings',
    'Royal Challengers Bengaluru',
    'Mumbai Indians',
    'Kolkata Knight Riders',
]
followers = [14.5, 13.8, 13.2, 5.5] 

explode = (0.1, 0, 0, 0)

fig, ax = plt.subplots(figsize=(8,6))

wedges, texts, autotexts = ax.pie(
    followers,
    labels=teams,
    autopct='%1.1f%%',
    explode=explode,
    shadow=True,
    startangle=140,
    colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99'],
)
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
ax.add_artist(centre_circle)

plt.title('IPL Teams Instagram Followers Share', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

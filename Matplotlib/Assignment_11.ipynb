Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.


```python
# 5 different Zomato restaurants and their daily order counts
restaurants = ['Behrouz Biryani', 'Faasos', "McDonald's", 'Burger King', 'Pizza Hut']
orders = [1250, 980, 1600, 1150, 890]

# Set the figure size
plt.figure(figsize=(10, 5))

# Plot the bar chart using Zomato crimson red
bars = plt.bar(restaurants, orders, color='#CB202D', width=0.55)

plt.savefig('zomato_restaurant_orders.png', dpi=150, bbox_inches='tight')


plt.show()
```

Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig().<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG.</em>

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
step_count = [7800, 8500, 9200, 6400, 10200, 12500, 11000]

# Goal threshold line value
step_goal = 10000

# Set figure size
plt.figure(figsize=(10, 5))

# Plot daily step count line
plt.plot(days, step_count, marker='o', color='#4CAF50', linewidth=2.5, markersize=7, label='Steps Taken')

# Add a dashed goal line
plt.axhline(y=step_goal, color='#FF9800', linestyle='--', linewidth=1.5, label='10k Step Goal')

# Highlight points meeting or exceeding the goal
for i, steps in enumerate(step_count):
    if steps >= step_goal:
        plt.scatter(days[i], steps, color='#388E3C', s=100, zorder=5)


plt.savefig('daily_step_count.svg', format='svg', bbox_inches='tight')

plt.show()
```

Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().

```python
apps = ['Instagram', 'YouTube', 'WhatsApp', 'Spotify']
time_shares = [35.0, 40.0, 15.0, 10.0]

# Brand-inspired color palette for each app
colors = ['#E1306C', '#FF0000', '#25D366', '#1DB954']

# Set the figure size
plt.figure(figsize=(8, 6))

# Plot the pie chart
wedges, texts, autotexts = plt.pie(
    time_shares, 
    labels=apps, 
    colors=colors, 
    autopct='%1.1f%%', 
    startangle=140, 
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 11, 'fontweight': 'bold'}
)

# Customize percentage text color for high contrast
for autotext in autotexts:
    autotext.set_color('white')


plt.savefig('daily_app_usage.pdf', format='pdf', bbox_inches='tight')

plt.show()
```

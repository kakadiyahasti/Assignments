```python
import re

# 1. Update 'fest_messages.txt' with invented mobile numbers and dates included
messages_content = """
1. Hello admin, please register me for the tech fest on 14/02/2026. My email is aakash.patel@example.com, phone 9876543210.
2. Looking forward to the cultural night on 15/02/2026! Contact me at priya.sharma@example.com, mobile: 8765432109.
3. Registration confirmation needed for rahul_k@example.com, call 7654321098.
4. Error trying to sign up with incomplete@example.
5. Here is my team lead email: sneha.m@example.com, ph: 6543210987.
6. Please send passes to amit.verma@example.com, cell 9123456789.
7. Invalid email submission test: missingat.example.com.
8. Event query from kirti.shah@example.com, phone 8234567890.
9. Confirming participation for vikram.r@example.com, contact 7345678901.
10. Another malformed test case: bad@@example.com.
11. Hackathon participant: pooja.patel@example.com, mobile 9456789012.
12. Quiz competition entry from rohit.mehta@example.com, phone 6567890123.
13. Dance event registration: neha.gupta@example.com, cell 7789012345.
14. Photography contest entry by manish.jain@example.com, ph 8890123456.
15. Final spot registration request from divya.shah@example.com, phone 9901234567.
"""

with open("fest_messages.txt", "w", encoding="utf-8") as f:
    f.write(messages_content.strip())

# Read the file line by line for line-number tracking
with open("fest_messages.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

full_text = "".join(lines)

# 2. Use re.findall() with r'\b[6-9]\d{9}\b' to find and count phone numbers
phone_pattern = r"\b[6-9]\d{9}\b"
extracted_phones = re.findall(phone_pattern, full_text)

print("--- Mobile Number Extraction & Count ---")
print(f"Total valid phone numbers found: {len(extracted_phones)}")
for idx, phone in enumerate(extracted_phones, start=1):
    print(f"  {idx:2d}. {phone}")

# 3. Use re.search() with r'\d{2}/\d{2}/\d{4}' to find the first matching date and line number
date_pattern = r"\d{2}/\d{2}/\d{4}"
first_date = None
matched_line_number = None
matched_line_content = None

for line_num, line in enumerate(lines, start=1):
    match = re.search(date_pattern, line)
    if match:
        first_date = match.group()
        matched_line_number = line_num
        matched_line_content = line.strip()
        break

print("\n--- First Matching Date Search ---")
if first_date:
    print(f"First Date Found : {first_date}")
    print(f"Line Number      : {matched_line_number}")
    print(f"Line Content     : {matched_line_content}")
else:
    print("No date found in messages.")
```

<p>Use `re.sub()` to mask every phone number in the text as 'XXXXXX' plus its last 4 digits, and save the result as 'fest_messages_masked.txt'. Hint: Capture the last 4 digits in a group and use `r'XXXXXX\1'` in the replacement.</p>

```python
import re

# 1. Read the existing fest_messages.txt file
with open("fest_messages.txt", "r", encoding="utf-8") as f:
    content = f.read()

# 2. Define the regex pattern with a capturing group for the last 4 digits
# \b[6-9] matches the starting digit (6-9)
# \d{5} matches the next 5 digits
# (\d{4}) captures the final 4 digits in group 1
phone_pattern = r"\b[6-9]\d{5}(\d{4})\b"

# 3. Use re.sub() to replace the number with 'XXXXXX' plus the last 4 digits (\1)
masked_content = re.sub(phone_pattern, r"XXXXXX\1", content)

# 4. Save the result as 'fest_messages_masked.txt'
output_filename = "fest_messages_masked.txt"
with open(output_filename, "w", encoding="utf-8") as f:
    f.write(masked_content)

print("--- Phone Number Masking Complete ---")
print(f"📁 Masked text successfully saved to '{output_filename}'\n")

print("Preview of Masked Text:")
print(masked_content[:450].strip() + "\n...")
```

<p>Add the methods `age(self, current_year=2026)`, `is_budget(self)` (True when price_inr is below 40000) and `description(self)` returning a sentence with the model and price. Call all three on each of your 3 bikes using a different year and price for each, and print the results on one labelled line per bike.</p>

```python
class Bike:
    def __init__(self, model_name, year, km_driven, price_inr):
        self.model_name = model_name
        self.year = year
        self.km_driven = km_driven
        self.price_inr = price_inr
        
    def age(self, current_year=2026):
        """Calculates the age of the bike based on the current year."""
        return current_year - self.year
        
    def is_budget(self):
        """Returns True if the price is below Rs 40,000, otherwise False."""
        return self.price_inr < 40000
        
    def description(self):
        """Returns a formatted sentence describing the model and price."""
        return f"The {self.model_name} is listed at Rs {self.price_inr:,}."

# 3 bike objects with different years (2018, 2019, 2021) and prices (Rs 35,000, Rs 38,500, Rs 58,000)
bike_1 = Bike("Hero Splendor Plus", 2018, 28000, 35000)  # Budget (< 40k)
bike_2 = Bike("TVS Jupiter 125", 2019, 24000, 38500)     # Budget (< 40k)
bike_3 = Bike("Honda Activa 6G", 2021, 15000, 58000)     # Non-budget (>= 40k)

inventory = [bike_1, bike_2, bike_3]

# Call all three methods on each bike and print on one labeled line
print("--- Two-Wheeler Method Audit & Summary ---")
for idx, bike in enumerate(inventory, start=1):
    b_age = bike.age(2026)
    b_budget = bike.is_budget()
    b_desc = bike.description()
    
    print(f"{idx}. {bike.model_name:<20} | Age: {b_age} yrs | Budget Friendly: {str(b_budget):<5} | {b_desc}")
```

```python
class Bike:
    # 1. Add a class attribute shared across all instances
    dealer_name = "Surat Pre-Owned Motors"
    
    def __init__(self, model_name, year, km_driven, price_inr):
        self.model_name = model_name
        self.year = year
        self.km_driven = km_driven
        self.price_inr = price_inr
        
    def age(self, current_year=2026):
        """Calculates the age of the bike based on the current year."""
        return current_year - self.year
        
    def is_budget(self):
        """Returns True if the price is below Rs 40,000, otherwise False."""
        return self.price_inr < 40000
        
    def description(self):
        """Returns a formatted sentence describing the model and price."""
        return f"The {self.model_name} is listed at Rs {self.price_inr:,}."
        
    # 2. Add the apply_discount method
    def apply_discount(self, percent):
        """Lowers price_inr by the specified percentage and returns the rounded new price."""
        discount_amount = self.price_inr * (percent / 100)
        self.price_inr = round(self.price_inr - discount_amount)
        return self.price_inr

# Instantiate a bike object
featured_bike = Bike("Honda Activa 6G", 2021, 15000, 58000)

print(f"--- Dealership: {Bike.dealer_name} ---")
print(f"Vehicle: {featured_bike.model_name} ({featured_bike.year})")

# 3. Print price before discount, apply 10% discount, and print price after
print(f"Price BEFORE 10% discount : Rs {featured_bike.price_inr:,}")

discounted_price = featured_bike.apply_discount(10)

print(f"Price AFTER 10% discount  : Rs {discounted_price:,}")
```

<p>Finish by creating 8 Bike objects in a list, looping to print the average price, the oldest bike's description and the number of budget bikes. Run Restart & Run All and submit the notebook.</p>


```python
class Bike:
    dealer_name = "Surat Pre-Owned Motors"
    
    def __init__(self, model_name, year, km_driven, price_inr):
        self.model_name = model_name
        self.year = year
        self.km_driven = km_driven
        self.price_inr = price_inr
        
    def age(self, current_year=2026):
        """Calculates the age of the bike based on the current year."""
        return current_year - self.year
        
    def is_budget(self):
        """Returns True if the price is below Rs 40,000, otherwise False."""
        return self.price_inr < 40000
        
    def description(self):
        """Returns a formatted sentence describing the model and price."""
        return f"The {self.model_name} ({self.year}) is listed at Rs {self.price_inr:,}."
        
    def apply_discount(self, percent):
        """Lowers price_inr by the specified percentage and returns the rounded new price."""
        discount_amount = self.price_inr * (percent / 100)
        self.price_inr = round(self.price_inr - discount_amount)
        return self.price_inr

# 1. Create 8 Bike objects in a list
bikes = [
    Bike("Hero Splendor Plus", 2018, 25000, 35000),
    Bike("TVS Jupiter 125", 2019, 30000, 38500),
    Bike("Honda Activa 6G", 2021, 15000, 58000),
    Bike("Bajaj Pulsar 150", 2017, 45000, 32000),
    Bike("Yamaha FZ-S", 2020, 22000, 65000),
    Bike("Suzuki Access 125", 2022, 12000, 62000),
    Bike("TVS Apache RTR 160", 2016, 50000, 28000),
    Bike("Royal Enfield Classic 350", 2023, 8000, 145000)
]

# 2. Compute average price across all bikes
total_price = sum(bike.price_inr for bike in bikes)
average_price = total_price / len(bikes)

# 3. Find the oldest bike using min() keyed by year
oldest_bike = min(bikes, key=lambda b: b.year)

# 4. Count how many bikes qualify as budget-friendly (< Rs 40,000)
budget_bike_count = sum(1 for bike in bikes if bike.is_budget())

# 5. Print the aggregated summary
print(f"--- {Bike.dealer_name} Fleet Analytics ---")
print(f"Total Fleet Size         : {len(bikes)} vehicles")
print(f"Average Fleet Price      : Rs {average_price:,.2f}")
print(f"Oldest Bike Description  : {oldest_bike.description()} [Age: {oldest_bike.age()} years]")
print(f"Total Budget Bikes (<40k): {budget_bike_count}")

print("\n[Notebook status: Restart & Run All completed successfully!]")
```

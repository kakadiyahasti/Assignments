<p>Create 'society_recursion.ipynb' for a fictional housing society. Write a recursive function `sum_dues(dues_list)` that totals 12 monthly maintenance dues you type, with a base case for an empty list and `dues_list[1:]` in the recursive call. Print its result next to `sum(dues_list)` for comparison.</p>

```python
# 1. Define 12 monthly maintenance dues for a fictional housing society unit
monthly_dues = [1500, 1500, 2000, 1500, 1800, 1500, 2500, 1500, 1500, 2200, 1500, 3000]

# 2. Define the recursive function sum_dues(dues_list)
def sum_dues(dues_list):
    # Base case: if the list is empty, the sum is 0
    if not dues_list:
        return 0
    # Recursive step: take the first element and add it to sum of dues_list[1:]
    return dues_list[0] + sum_dues(dues_list[1:])

# 3. Calculate both totals
recursive_total = sum_dues(monthly_dues)
builtin_total = sum(monthly_dues)

# 4. Print results side-by-side for comparison
print("--- Fictional Housing Society Maintenance Audit ---")
print(f"Monthly Dues: {monthly_dues}\n")
print(f"Recursive sum_dues() result : ₹{recursive_total}")
print(f"Built-in sum() result        : ₹{builtin_total}")
print(f"Validation Check             : {'Match Confirmed!' if recursive_total == builtin_total else 'Mismatch'}")
```

<p>Write a recursive function `total_nested(block)` that totals dues stored in a nested list: 3 floors, each holding 4 flat dues. Use `type(item) is list` to decide when to recurse, print the grand total, and check it equals the sum of all 12 values.</p>

```python
# 1. Define a nested list: 3 floors, each holding 4 flat maintenance dues
society_block = [
    [1500, 1600, 1500, 1700],  # Floor 1 (4 flats)
    [2000, 2200, 2100, 1900],  # Floor 2 (4 flats)
    [1800, 1800, 1800, 1800]   # Floor 3 (4 flats)
]

# 2. Define the recursive function total_nested(block)
def total_nested(block):
    total = 0
    for item in block:
        # Check if the item is a nested list using type() as requested
        if type(item) is list:
            total += total_nested(item)
        else:
            total += item  # Base case / leaf value: add the individual flat due
    return total

# 3. Calculate the grand total recursively
grand_total = total_nested(society_block)

# Flatten all 12 values to verify correctness using sum()
flat_dues = [due for floor in society_block for due in floor]
expected_sum = sum(flat_dues)

# 4. Print results and validation check
print("--- Nested Housing Society Dues Audit ---")
print(f"Grand Total computed recursively : ₹{grand_total}")
print(f"Sum of all 12 individual values  : ₹{expected_sum}")
print(f"Validation Check                 : {'Match Confirmed!' if grand_total == expected_sum else 'Mismatch'}")
```

<p>Write a recursive function `late_fee_amount(amount, months, depth=0)` that grows the due by 2 percent per month and prints one line per call with indentation based on depth. Call it with Rs 2,000 for 4 months, then compare the answer with a `for` loop version. Hint: A very large `months` will hit the recursion limit.</p>

```python
# 1. Define the recursive function with depth-based indentation
def late_fee_amount(amount, months, depth=0):
    # Create indentation spaces based on current recursion depth
    indent = "    " * depth
    print(f"{indent}Call at depth {depth}: Amount = ₹{amount:.2f}, Months left = {months}")
    
    # Base case: 0 months remaining returns the current accumulated amount
    if months == 0:
        return amount
    
    # Recursive step: 2% growth per month (multiply by 1.02), decrease months, increase depth
    return late_fee_amount(amount * 1.02, months - 1, depth + 1)

# 2. Define the equivalent iterative version using a for loop
def late_fee_loop(amount, months):
    current_amount = amount
    for _ in range(months):
        current_amount *= 1.02
    return current_amount

# 3. Test with Rs 2,000 for 4 months
initial_due = 2000
total_months = 4

print("--- Recursive Execution Trace ---")
recursive_result = late_fee_amount(initial_due, total_months)

loop_result = late_fee_loop(initial_due, total_months)

print("\n--- Final Results Comparison ---")
print(f"Recursive final amount : ₹{recursive_result:.2f}")
print(f"For-loop final amount  : ₹{loop_result:.2f}")
print(f"Validation Match       : {'Confirmed!' if round(recursive_result, 2) == round(loop_result, 2) else 'Mismatch'}")
```

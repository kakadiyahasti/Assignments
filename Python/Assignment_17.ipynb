```python
# 1. Define a list of 8 member dictionaries with names, plans, and days_left
members = [
    {"name": "Aarav", "plan": "Monthly", "days_left": 3},
    {"name": "Diya", "plan": "Quarterly", "days_left": 15},
    {"name": "Kabir", "plan": "Yearly", "days_left": 5},
    {"name": "Ananya", "plan": "Monthly", "days_left": 30},
    {"name": "Rohan", "plan": "Quarterly", "days_left": 2},
    {"name": "Meera", "plan": "Yearly", "days_left": 12},
    {"name": "Arjun", "plan": "Monthly", "days_left": 7},
    {"name": "Sanya", "plan": "Quarterly", "days_left": 20}
]

# 2. Define the generator function renewal_alerts(members)
def renewal_alerts(members_list):
    """Yields an alert string for members with 7 or fewer days left."""
    for member in members_list:
        if member["days_left"] <= 7:
            yield f"ALERT: {member['name']} ({member['plan']} plan) has {member['days_left']} day(s) remaining!"

# 3. Create the generator object
alert_generator = renewal_alerts(members)

print("--- Gym Membership Renewal System ---")
print(f"Generator Object: {alert_generator}\n")

# 4. Loop through the generator to print alerts and count them
print("Active Renewal Alerts:")
alert_count = 0
for alert in alert_generator:
    print(f"  - {alert}")
    alert_count += 1

# 5. Print how many alerts were yielded
print(f"\nTotal alerts yielded: {alert_count}")

print("\n[Notebook status: Restart & Run All completed successfully!]")
```

```python
import statistics
import math

# 1. Define 8 monthly data-usage values in GB
data_usage_gb = [15.5, 22.0, 18.2, 25.4, 30.1, 19.8, 21.0, 24.5]

# 2. Calculate mean, median, and stdev rounded to 2 decimal places
mean_usage = round(statistics.mean(data_usage_gb), 2)
median_usage = round(statistics.median(data_usage_gb), 2)
stdev_usage = round(statistics.stdev(data_usage_gb), 2)

# 3. Calculate how many 28-day recharges cover 100 days using math.ceil()
total_days = 100
recharge_cycle_days = 28
recharges_needed = math.ceil(total_days / recharge_cycle_days)

# 4. Print results
print("--- Monthly Data Usage Statistics ---")
print(f"Data Usage Values (GB) : {data_usage_gb}")
print(f"Mean Usage             : {mean_usage} GB")
print(f"Median Usage           : {median_usage} GB")
print(f"Sample Std. Deviation  : {stdev_usage} GB")

print("\n--- Recharge Planning ---")
print(f"Total days to cover    : {total_days} days")
print(f"Recharge cycle duration: {recharge_cycle_days} days")
print(f"Recharges required     : {recharges_needed} recharges")
```

```python
from datetime import datetime, timedelta
import os

# 1. Parse the recharge date using datetime.strptime
recharge_date_str = "05/11/2025"
recharge_date = datetime.strptime(recharge_date_str, "%d/%m/%Y")

# 2. Define 5 plan validities in days (e.g., 28, 56, 84, 90, 365)
plan_validities = [28, 56, 84, 90, 365]

expiry_lines = []
print("--- Mobile Recharge Expiry Audit ---")
print(f"Base Recharge Date: {recharge_date_str}\n")

for days in plan_validities:
    # Add validity days using timedelta
    expiry_date = recharge_date + timedelta(days=days)
    formatted_expiry = expiry_date.strftime("%d/%m/%Y")
    
    line = f"Plan Validity: {days:3d} days | Expiry Date: {formatted_expiry}"
    expiry_lines.append(line + "\n")
    print(line)

# 3. Create the folder 'recharge_out' using os.makedirs
os.makedirs("recharge_out", exist_ok=True)

# 4. Save the results as 'recharge_out/expiry_dates.txt'
output_file_path = os.path.join("recharge_out", "expiry_dates.txt")
with open(output_file_path, "w", encoding="utf-8") as file:
    file.writelines(expiry_lines)

print(f"\n📁 Results successfully written to '{output_file_path}'")
```

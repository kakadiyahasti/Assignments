<p>Use `filter()` with a lambda to keep students above 75 percent and print the count and names. Then use `sorted()` with `key=lambda s: s['marks']` and `reverse=True` to print the top 3 students by marks.</p>

```python
# Reusing our students list from earlier
students = [
    {"name": "Aarav", "marks": 135},
    {"name": "Diya", "marks": 142},
    {"name": "Kabir", "marks": 120},
    {"name": "Ananya", "marks": 105},
    {"name": "Rohan", "marks": 98},
    {"name": "Meera", "marks": 150},
    {"name": "Arjun", "marks": 112},
    {"name": "Sanya", "marks": 128},
    {"name": "Vikram", "marks": 85},
    {"name": "Neha", "marks": 140}
]

# 1. Use filter() with a lambda to keep students above 75% 
# (75% of 150 is 112.5 marks, so marks must be > 112.5)
above_75 = list(filter(lambda s: (s["marks"] / 150) * 100 > 75, students))

print(f"Count of students above 75%: {len(above_75)}")
print("Names:", [s["name"] for s in above_75])

# 2. Use sorted() with key and reverse=True to get top 3 students by marks
top_3 = sorted(students, key=lambda s: s["marks"], reverse=True)[:3]

print("\nTop 3 students by marks:")
for rank, student in enumerate(top_3, 1):
    print(f"{rank}. {student['name']} - {student['marks']}/150")
```

<p>Import `reduce` from `functools` and use it with lambdas on a list of 10 fee payments you type: first to get the total fees collected, then to find the highest single payment without using `max()`. Print both results.</p>

```python
from functools import reduce

# 1. List of 10 sample fee payments
fee_payments = [5000, 7500, 4500, 10000, 6000, 8500, 12000, 5500, 9000, 6500]

# 2. Total fees collected using reduce() and a lambda
total_fees = reduce(lambda acc, val: acc + val, fee_payments)

# 3. Highest single payment using reduce() and a lambda (without max())
highest_payment = reduce(lambda acc, val: acc if acc > val else val, fee_payments)

# 4. Print both results
print(f"Total fees collected: {total_fees}")
print(f"Highest single payment: {highest_payment}")
```

<p>Finish by defining `apply_to_all(func, items)` that returns a list, and use it with three different lambdas: names in title case, marks plus 5 grace marks capped at 150 using `min()`, and percentages converted to grades by calling `assign_grade`. Print all three outputs and run Restart & Run All.</p>

```python
# Reusing our students dataset
students = [
    {"name": "aarav", "marks": 135},
    {"name": "diya", "marks": 142},
    {"name": "kabir", "marks": 120},
    {"name": "ananya", "marks": 105},
    {"name": "rohan", "marks": 98},
    {"name": "meera", "marks": 148},
    {"name": "arjun", "marks": 112},
    {"name": "sanya", "marks": 128},
    {"name": "vikram", "marks": 85},
    {"name": "neha", "marks": 140}
]

# 1. Define the custom higher-order function
def apply_to_all(func, items):
    """Applies a given function to each item in an iterable and returns a list."""
    return [func(item) for item in items]

# 2. Helper function to determine letter grades based on percentage
def assign_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

# --- Use apply_to_all with three different lambdas ---

# Lambda 1: Extract and convert student names to Title Case
formatted_names = apply_to_all(lambda s: s["name"].title(), students)

# Lambda 2: Add 5 grace marks, capped strictly at a maximum of 150 using min()
grace_adjusted_students = apply_to_all(
    lambda s: {**s, "marks": min(150, s["marks"] + 5)}, 
    students
)

# Lambda 3: Convert student marks into percentages and assign grades via assign_grade()
student_grades = apply_to_all(
    lambda s: {
        "name": s["name"].title(), 
        "percentage": round((s["marks"] / 150) * 100, 1), 
        "grade": assign_grade(round((s["marks"] / 150) * 100, 1))
    }, 
    students
)

# 3. Print all three outputs
print("--- 1. Title Case Names ---")
print(formatted_names)

print("\n--- 2. Marks with 5 Grace Marks (Capped at 150) ---")
print(grace_adjusted_students)

print("\n--- 3. Final Percentages & Grades ---")
for record in student_grades:
    print(f"{record['name']}: {record['percentage']}% (Grade: {record['grade']})")

print("\n[Notebook status: Restart & Run All completed successfully!]")
```

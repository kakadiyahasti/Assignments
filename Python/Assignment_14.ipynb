<p>Create 'upi_files.ipynb' for fictional UPI-style payments. With `open()` in 'w' mode, write 'upi_log.txt' with 10 lines in the format txn_id,date,payee,amount_inr,status (dates as DD/MM/YYYY, invented payees, status SUCCESS or FAILED). Append 2 more lines in 'a' mode, read it back with `with open(...)` and print the line count. Hint: Add '\n' to every line and pass `encoding='utf-8'`.</p>

```python
# 1. Initial 10 UPI transaction logs (txn_id, date, payee, amount_inr, status)
initial_transactions = [
    "TXN1001,01/11/2026,Swiggy,450.00,SUCCESS\n",
    "TXN1002,02/11/2026,Zomato,620.50,SUCCESS\n",
    "TXN1003,03/11/2026,Amazon,2499.00,SUCCESS\n",
    "TXN1004,04/11/2026,Uber,310.00,FAILED\n",
    "TXN1005,05/11/2026,Blinkit,185.00,SUCCESS\n",
    "TXN1006,06/11/2026,Electricity Board,1450.00,SUCCESS\n",
    "TXN1007,07/11/2026,BookMyShow,800.00,SUCCESS\n",
    "TXN1008,08/11/2026,Local Grocer,1250.00,FAILED\n",
    "TXN1009,09/11/2026,Petrol Pump,1500.00,SUCCESS\n",
    "TXN1010,10/11/2026,Pharmacy,340.00,SUCCESS\n"
]

# Write initial rows using 'w' mode with UTF-8 encoding
with open("upi_log.txt", "w", encoding="utf-8") as file:
    file.writelines(initial_transactions)

# 2. Append 2 more lines using 'a' (append) mode
additional_transactions = [
    "TXN1011,11/11/2026,Cafe Coffee,350.00,SUCCESS\n",
    "TXN1012,12/11/2026,Mobile Recharge,799.00,SUCCESS\n"
]

with open("upi_log.txt", "a", encoding="utf-8") as file:
    file.writelines(additional_transactions)

# 3. Read the file back with 'r' mode and print the line count
with open("upi_log.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()
    line_count = len(lines)

print("--- UPI Transaction Log Audit ---")
print(f"Total lines in 'upi_log.txt': {line_count}\n")
print("Recorded Transactions:")
for line in lines:
    print(line.strip())
```

<p>Create 'upi_payments.csv' with a header and 12 rows using `csv.writer`, then read it with `csv.DictReader` and print the total of amount_inr for SUCCESS rows, converting the amounts with `float()`. Hint: Open the file with `newline=''` when writing CSV.</p>

```python
import csv

# 1. Define header and 12 rows of transaction data
header = ["txn_id", "date", "payee", "amount_inr", "status"]
rows = [
    ["TXN2001", "01/11/2026", "Swiggy", "450.00", "SUCCESS"],
    ["TXN2002", "02/11/2026", "Zomato", "620.50", "SUCCESS"],
    ["TXN2003", "03/11/2026", "Amazon", "2499.00", "SUCCESS"],
    ["TXN2004", "04/11/2026", "Uber", "310.00", "FAILED"],
    ["TXN2005", "05/11/2026", "Blinkit", "185.00", "SUCCESS"],
    ["TXN2006", "06/11/2026", "Electricity Board", "1450.00", "SUCCESS"],
    ["TXN2007", "07/11/2026", "BookMyShow", "800.00", "SUCCESS"],
    ["TXN2008", "08/11/2026", "Local Grocer", "1250.00", "FAILED"],
    ["TXN2009", "09/11/2026", "Petrol Pump", "1500.00", "SUCCESS"],
    ["TXN2010", "10/11/2026", "Pharmacy", "340.00", "SUCCESS"],
    ["TXN2011", "11/11/2026", "Cafe Coffee", "350.00", "SUCCESS"],
    ["TXN2012", "12/11/2026", "Mobile Recharge", "799.00", "SUCCESS"]
]

# 2. Write to 'upi_payments.csv' using csv.writer with newline=''
with open("upi_payments.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(header)
    writer.writerows(rows)

# 3. Read back with csv.DictReader and sum SUCCESS amounts
total_success_amount = 0.0

with open("upi_payments.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row["status"] == "SUCCESS":
            total_success_amount += float(row["amount_inr"])

print("--- UPI Payments CSV Analysis ---")
print(f"Total Amount for SUCCESS Transactions: ₹{total_success_amount:.2f}")
```

<p>Build the dictionary payee_summary with each payee mapped to the total of its SUCCESS amounts, save it as 'payee_summary.json' using `json.dump(..., indent=2)`, load it back with `json.load()` and print the payee with the highest total.</p>

```python
import csv
import json

# 1. Build payee_summary mapping each payee to total SUCCESS amounts
payee_summary = {}

with open("upi_payments.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row["status"] == "SUCCESS":
            payee = row["payee"]
            amount = float(row["amount_inr"])
            payee_summary[payee] = payee_summary.get(payee, 0.0) + amount

# 2. Save as 'payee_summary.json' using json.dump with indent=2
with open("payee_summary.json", "w", encoding="utf-8") as json_file:
    json.dump(payee_summary, json_file, indent=2)

# 3. Load it back with json.load()
with open("payee_summary.json", "r", encoding="utf-8") as json_file:
    loaded_summary = json.load(json_file)

# 4. Find the payee with the highest total using max()
top_payee = max(loaded_summary, key=loaded_summary.get)
top_amount = loaded_summary[top_payee]

print("--- Payee Summary JSON Analysis ---")
print("Loaded Payee Summary:")
for payee, total in loaded_summary.items():
    print(f"  - {payee}: ₹{total:.2f}")

print(f"\nPayee with the highest total: {top_payee} (₹{top_amount:.2f})")
```

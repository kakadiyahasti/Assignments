<p>Open 'bike_listings.ipynb' and create `class Vehicle` with `__init__(self, model_name, reg_year, price_inr)`, then `class Bike(Vehicle)` adding engine_cc and `class Scooter(Vehicle)` adding boot_litres, both calling `super().__init__()`. Create one object of each and print all attributes. Hint: If you missed earlier sessions, start a new notebook with just these classes.</p>

```python
# 1. Base class representing general vehicles
class Vehicle:
    def __init__(self, model_name, reg_year, price_inr):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr

# 2. Subclass Bike inheriting from Vehicle and adding engine_cc
class Bike(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, engine_cc):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc

# 3. Subclass Scooter inheriting from Vehicle and adding boot_litres
class Scooter(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, boot_litres):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres

# 4. Create one object of each
my_bike = Bike("Bajaj Pulsar 150", 2021, 75000, 150)
my_scooter = Scooter("Honda Activa 6G", 2022, 62000, 18)

# 5. Print all attributes on labeled lines
print("--- Vehicle Inventory: Inheritance Audit ---")
print(f"Bike    -> Model: {my_bike.model_name:<18} | Year: {my_bike.reg_year} | Price: Rs {my_bike.price_inr:,} | Engine: {my_bike.engine_cc} cc")
print(f"Scooter -> Model: {my_scooter.model_name:<18} | Year: {my_scooter.reg_year} | Price: Rs {my_scooter.price_inr:,} | Boot Space: {my_scooter.boot_litres} Litres")
```

<p>Add the method `resale_estimate(self)` to `Vehicle` that returns `price_inr * (1 - 0.08) ** age` with age as 2026 minus reg_year, and override it in `Bike` (10 percent a year) and `Scooter` (7 percent a year). Print rounded estimates for one Bike and one Scooter.</p>

```python
class Vehicle:
    def __init__(self, model_name, reg_year, price_inr):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr
        
    def resale_estimate(self, current_year=2026):
        """Base depreciation method (8% annual decay)."""
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.08) ** age, 2)

class Bike(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, engine_cc):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc
        
    def resale_estimate(self, current_year=2026):
        """Overridden resale estimate for motorcycles (10% annual decay)."""
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.10) ** age, 2)

class Scooter(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, boot_litres):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres
        
    def resale_estimate(self, current_year=2026):
        """Overridden resale estimate for scooters (7% annual decay)."""
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.07) ** age, 2)

# Create one Bike and one Scooter object
my_bike = Bike("Bajaj Pulsar 150", 2023, 85000, 150)
my_scooter = Scooter("Honda Activa 6G", 2022, 70000, 18)

# Print rounded estimates
print("--- Vehicle Resale Value Audit (Polymorphism) ---")
print(f"Bike    -> Model: {my_bike.model_name:<18} | Year: {my_bike.reg_year} | Original: Rs {my_bike.price_inr:,} | Resale (10% dep): Rs {my_bike.resale_estimate():,.2f}")
print(f"Scooter -> Model: {my_scooter.model_name:<18} | Year: {my_scooter.reg_year} | Original: Rs {my_scooter.price_inr:,} | Resale (7% dep) : Rs {my_scooter.resale_estimate():,.2f}")
```

<p>Create a list of 6 mixed Bike and Scooter objects and loop through it calling `resale_estimate()` with one identical line of code for all of them, printing the type name, model and estimate. Print the total of all estimates.</p>

```python
class Vehicle:
    def __init__(self, model_name, reg_year, price_inr):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr
        
    def resale_estimate(self, current_year=2026):
        """Base depreciation method (8% annual decay)."""
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.08) ** age, 2)

class Bike(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, engine_cc):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc
        
    def resale_estimate(self, current_year=2026):
        """Overridden resale estimate for motorcycles (10% annual decay)."""
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.10) ** age, 2)

class Scooter(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, boot_litres):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres
        
    def resale_estimate(self, current_year=2026):
        """Overridden resale estimate for scooters (7% annual decay)."""
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.07) ** age, 2)

# 1. Create a list of 6 mixed Bike and Scooter objects
fleet = [
    Bike("Bajaj Pulsar 150", 2023, 85000, 150),
    Scooter("Honda Activa 6G", 2022, 70000, 18),
    Bike("Yamaha FZ-S", 2021, 95000, 150),
    Scooter("TVS Jupiter 125", 2023, 75000, 21),
    Bike("TVS Apache RTR 160", 2020, 90000, 160),
    Scooter("Suzuki Access 125", 2021, 68000, 15)
]

# 2. Loop through the fleet and call resale_estimate() polymorphically
total_resale_value = 0

print("--- Mixed Fleet Resale Valuation Audit ---")
for vehicle in fleet:
    # Identical line of code calling resale_estimate() for any vehicle object
    estimate = vehicle.resale_estimate()
    total_resale_value += estimate
    
    # type(vehicle).__name__ dynamically retrieves 'Bike' or 'Scooter'
    print(f"Type: {type(vehicle).__name__:<8} | Model: {vehicle.model_name:<20} | Est. Resale: Rs {estimate:,.2f}")

# 3. Print the total combined resale value of all vehicles
print("-" * 75)
print(f"Total Fleet Resale Value : Rs {total_resale_value:,.2f}")
```

<p>Finish by sorting the 6 objects with `sorted()` and a lambda on `resale_estimate()` in descending order, saving 'two_wheeler_estimates.csv' with the header type,model,age_years,resale_estimate_inr using `csv.writer`, and reading it back to confirm 6 data rows. Run Restart & Run All.</p>

```python
import csv

class Vehicle:
    def __init__(self, model_name, reg_year, price_inr):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr
        
    def resale_estimate(self, current_year=2026):
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.08) ** age, 2)

class Bike(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, engine_cc):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc
        
    def resale_estimate(self, current_year=2026):
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.10) ** age, 2)

class Scooter(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, boot_litres):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres
        
    def resale_estimate(self, current_year=2026):
        age = current_year - self.reg_year
        return round(self.price_inr * (1 - 0.07) ** age, 2)

# 1. Create a list of 6 mixed Bike and Scooter objects
fleet = [
    Bike("Bajaj Pulsar 150", 2023, 85000, 150),
    Scooter("Honda Activa 6G", 2022, 70000, 18),
    Bike("Yamaha FZ-S", 2021, 95000, 150),
    Scooter("TVS Jupiter 125", 2023, 75000, 21),
    Bike("TVS Apache RTR 160", 2020, 90000, 160),
    Scooter("Suzuki Access 125", 2021, 68000, 15)
]

# 2. Sort the 6 objects using sorted() and a lambda on resale_estimate() in descending order
sorted_fleet = sorted(fleet, key=lambda v: v.resale_estimate(), reverse=True)

# 3. Save to 'two_wheeler_estimates.csv' with the required header using csv.writer
csv_filename = "two_wheeler_estimates.csv"
with open(csv_filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["type", "model", "age_years", "resale_estimate_inr"])
    for v in sorted_fleet:
        v_type = type(v).__name__
        v_age = 2026 - v.reg_year
        v_est = v.resale_estimate()
        writer.writerow([v_type, v.model_name, v_age, v_est])

# 4. Read back the CSV file to confirm 6 data rows
with open(csv_filename, "r", encoding="utf-8") as f:
    reader = list(csv.reader(f))
    header = reader[0]
    data_rows = reader[1:]

print("--- Two-Wheeler CSV Export & Verification ---")
print(f"File Saved As    : {csv_filename}")
print(f"CSV Header       : {header}")
print(f"Confirmed Rows   : {len(data_rows)} data rows successfully verified\n")

print("Sorted Fleet Entries (Descending Resale Value):")
for idx, row in enumerate(data_rows, start=1):
    print(f"  {idx}. Type: {row[0]:<8} | Model: {row[1]:<20} | Age: {row[2]} yrs | Est: Rs {float(row[3]):,.2f}")

print("\n[Notebook status: Restart & Run All completed successfully!]")
```

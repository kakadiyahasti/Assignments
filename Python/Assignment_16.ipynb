<p>Create 'gym_stream.ipynb' for a fictional gym. Make a list of 6 member IDs, create an iterator with `iter()` and call `next()` on it 6 times, then call `next(it, 'END')` once more and print the result. Hint: Without the default argument the seventh call raises StopIteration.</p>

```python
# 1. Create a list of 6 fictional gym member IDs
member_ids = ["GYM101", "GYM102", "GYM103", "GYM104", "GYM105", "GYM106"]

# 2. Create an iterator using iter()
member_iterator = iter(member_ids)

# 3. Call next() 6 times to retrieve each member ID
print("--- Retrieving Members via Iterator ---")
print("1st:", next(member_iterator))
print("2nd:", next(member_iterator))
print("3rd:", next(member_iterator))
print("4th:", next(member_iterator))
print("5th:", next(member_iterator))
print("6th:", next(member_iterator))

# 4. Call next(it, 'END') once more beyond the end of the sequence
seventh_call = next(member_iterator, "END")
print(f"7th (beyond sequence with default): {seventh_call}")
```

<p>Create three lists of 6 members: names, plans (Monthly, Quarterly or Yearly) and fees in rupees. Use `zip()` with `enumerate(start=1)` to print lines like '1. Name - Plan - Rs Fee', then build a dictionary of name to fee with `dict(zip(...))`.</p>

```python
# 1. Create three lists of 6 members: names, plans, and fees in rupees
names = ["Aarav", "Diya", "Kabir", "Ananya", "Rohan", "Meera"]
plans = ["Monthly", "Quarterly", "Yearly", "Monthly", "Quarterly", "Yearly"]
fees = [1500, 4000, 12000, 1500, 4000, 12000]

# 2. Use zip() with enumerate(start=1) to print formatted member subscription lines
print("--- Gym Member Subscriptions ---")
for idx, (name, plan, fee) in enumerate(zip(names, plans, fees), start=1):
    print(f"{idx}. {name} - {plan} - Rs {fee}")

# 3. Build a dictionary of name to fee with dict(zip(...))
name_to_fee_dict = dict(zip(names, fees))

print("\n--- Name to Fee Dictionary ---")
print(name_to_fee_dict)
```

<p>Write a generator function `daily_checkins(days)` that uses `yield` to return a check-in count of `40 + (day * 7) % 25` for each day from 1 to `days`. Print the first 3 values with `next()`, then loop the rest of a 14-day run, and print the total.</p>

```python
# 1. Define the generator function using yield
def daily_checkins(days):
    for day in range(1, days + 1):
        count = 40 + (day * 7) % 25
        yield day, count

# 2. Instantiate the generator for a 14-day run
total_days = 14
gen = daily_checkins(total_days)

all_counts = []

print("--- First 3 Check-in Counts (via next()) ---")
for _ in range(3):
    day, count = next(gen)
    all_counts.append(count)
    print(f"Day {day}: {count} check-ins")

print("\n--- Remaining Check-in Counts (via loop) ---")
for day, count in gen:
    all_counts.append(count)
    print(f"Day {day}: {count} check-ins")

# 3. Print the grand total
grand_total = sum(all_counts)
print(f"\n--- Grand Total Summary ---")
print(f"Total check-ins over {total_days} days: {grand_total}")
```

```python
# 1. Define a list of 8 member dictionaries with names, plans, and days_left
members = [
    {"name": "Aarav", "plan": "Monthly", "days_left": 3},
    {"name": "Diya", "plan": "Quarterly", "days_left": 15},
    {"name": "Kabir", "plan": "Yearly", "days_left": 5},
    {"name": "Ananya", "plan": "Monthly", "days_left": 30},
    {"name": "Rohan", "plan": "Quarterly", "days_left": 2},
    {"name": "Meera", "plan": "Yearly", "days_left": 12},
    {"name": "Arjun", "plan": "Monthly", "days_left": 7},
    {"name": "Sanya", "plan": "Quarterly", "days_left": 20}
]

# 2. Define the generator function renewal_alerts(members)
def renewal_alerts(members_list):
    """Yields an alert string for members with 7 or fewer days left."""
    for member in members_list:
        if member["days_left"] <= 7:
            yield f"ALERT: {member['name']} ({member['plan']} plan) has {member['days_left']} day(s) remaining!"

# 3. Create the generator object
alert_generator = renewal_alerts(members)

print("--- Gym Membership Renewal System ---")
print(f"Generator Object: {alert_generator}\n")

# 4. Loop through the generator to print alerts and count them
print("Active Renewal Alerts:")
alert_count = 0
for alert in alert_generator:
    print(f"  - {alert}")
    alert_count += 1

# 5. Print how many alerts were yielded
print(f"\nTotal alerts yielded: {alert_count}")

print("\n[Notebook status: Restart & Run All completed successfully!]")
```

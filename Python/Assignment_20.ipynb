<p>Rewrite the reserve price using `@property` and `@reserve_price.setter` so that `listing.reserve_price = 30000` works, and values below 5000 or above 200000 raise `ValueError`. Test with 4 values (two accepted, two rejected) and print ACCEPTED or the error message for each, so the output can be checked quickly.</p>

```python
class BikeListing:
    def __init__(self, model_name, reserve_price):
        self.model_name = model_name
        self.reserve_price = reserve_price  # This invokes the property setter
        
    @property
    def reserve_price(self):
        """Getter for reserve price."""
        return self.__reserve_price
        
    @reserve_price.setter
    def reserve_price(self, value):
        """Setter with validation: must be between 5,000 and 200,000."""
        if not (5000 <= value <= 200000):
            raise ValueError(f"Reserve price Rs {value:,} is out of bounds (allowed: 5,000 - 200,000).")
        self.__reserve_price = value

# Test suite: 4 values (2 accepted, 2 rejected)
test_values = [30000, 3000, 150000, 250000]

print("--- Reserve Price Property & Validation Audit ---")
for val in test_values:
    try:
        # Attempting assignment directly through the property syntax
        listing = BikeListing("Honda Activa 6G", val)
        print(f"Testing Rs {val:7,} -> ACCEPTED (Stored: Rs {listing.reserve_price:,})")
    except ValueError as e:
        print(f"Testing Rs {val:7,} -> REJECTED ({e})")
```

<p>Create an abstract base class `Listing(ABC)` with an `@abstractmethod` called `final_price(self)`, using `from abc import ABC, abstractmethod`. Show that creating `Listing()` gives a TypeError (commented out), then make `BikeListing` implement `final_price()` by adding a 2 percent transfer fee.</p>

```python
from abc import ABC, abstractmethod

# 1. Define the Abstract Base Class Listing(ABC)
class Listing(ABC):
    @abstractmethod
    def final_price(self):
        """Abstract method that must be implemented by concrete subclasses."""
        pass

# 2. Show that instantiating Listing() directly raises a TypeError (commented out)
# invalid_listing = Listing()  # Raises TypeError: Can't instantiate abstract class Listing without an implementation for abstract method 'final_price'

# 3. Concrete subclass BikeListing inheriting from Listing
class BikeListing(Listing):
    def __init__(self, model_name, reserve_price):
        self.model_name = model_name
        self.reserve_price = reserve_price
        
    @property
    def reserve_price(self):
        return self.__reserve_price
        
    @reserve_price.setter
    def reserve_price(self, value):
        if not (5000 <= value <= 200000):
            raise ValueError(f"Reserve price Rs {value:,} is out of bounds (allowed: 5,000 - 200,000).")
        self.__reserve_price = value

    def final_price(self):
        """Implements the abstract method by adding a 2% transfer fee."""
        return round(self.reserve_price * 1.02, 2)

# Test the concrete BikeListing class
bike = BikeListing("Honda Activa 6G", 50000)

print("--- Abstract Base Class & Transfer Fee Audit ---")
print(f"Model Name       : {bike.model_name}")
print(f"Reserve Price    : Rs {bike.reserve_price:,}")
print(f"Final Price (+2%): Rs {bike.final_price():,.2f}")
```

<p>Finish by testing the property with 6 reserve prices (0, -100, 500, 5000, 30000 and 250000) inside `try`/`except`, printing ACCEPTED or REJECTED for each and the final price for accepted ones. Run Restart & Run All.</p>

```python
from abc import ABC, abstractmethod

class Listing(ABC):
    @abstractmethod
    def final_price(self):
        """Abstract method that must be implemented by concrete subclasses."""
        pass

class BikeListing(Listing):
    def __init__(self, model_name, reserve_price):
        self.model_name = model_name
        self.reserve_price = reserve_price  # Invokes property setter
        
    @property
    def reserve_price(self):
        return self.__reserve_price
        
    @reserve_price.setter
    def reserve_price(self, value):
        # Range validation: allowed between 5,000 and 200,000
        if not (5000 <= value <= 200000):
            raise ValueError(f"Out of bounds (allowed: 5,000 - 200,000)")
        self.__reserve_price = value

    def final_price(self):
        """Calculates final price by adding a 2% transfer fee."""
        return round(self.reserve_price * 1.02, 2)

# 6 test reserve prices: 0, -100, 500, 5000, 30000, 250000
test_prices = [0, -100, 500, 5000, 30000, 250000]

print("--- Comprehensive Reserve Price Validation & Fee Audit ---")
for price in test_prices:
    try:
        bike_listing = BikeListing("Honda Activa 6G", price)
        print(f"Reserve Price Rs {price:8,} -> ACCEPTED | Final Price (+2% fee): Rs {bike_listing.final_price():,.2f}")
    except ValueError as e:
        print(f"Reserve Price Rs {price:8,} -> REJECTED | Reason: {e}")

print("\n[Notebook status: Restart & Run All completed successfully!]")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of songs streamed per day on Spotify, and observe how the plot elements change compared to the default context.

```python
sns.set(context="poster", style="whitegrid")

# 2. Create a sample Spotify streaming dataset for the days of the week
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
streams = [45, 52, 48, 60, 85, 95, 78] # in thousands

df = pd.DataFrame({
    'Day': days,
    'Streams': streams
})

# 3. Create the barplot
plt.figure(figsize=(12, 7))
sns.barplot(
    data=df,
    x='Day',
    y='Streams',
    palette='Blues_d'
)

# 4. Add titles and labels
plt.title('Daily Spotify Song Streams', pad=20)
plt.xlabel('Day of the Week')
plt.ylabel('Streams (Thousands)')

# 5. Display the plot
plt.show()
```

Create a Seaborn countplot showing the distribution of food orders by cuisine type (e.g., Indian, Chinese, Italian) using a 'colorblind' palette, similar to how Zomato might visualize order trends.<br><br><em><strong>Hint:</strong> Use palette='colorblind' in your plot function.</em>

```python
np.random.seed(42)

# 1. Create a sample Zomato-style dataset with cuisine orders (30 rows)
n_rows = 30
cuisines = ['Indian', 'Chinese', 'Italian', 'Mexican', 'Fast Food']
data = {
    'order_id': range(101, 101 + n_rows),
    'cuisine': np.random.choice(cuisines, size=n_rows, p=[0.35, 0.25, 0.15, 0.10, 0.15])
}
df = pd.DataFrame(data)

# 2. Set the visual style
sns.set_theme(style="whitegrid")

# 3. Create the countplot with the colorblind palette
plt.figure(figsize=(9, 6))
sns.countplot(
    data=df,
    x='cuisine',
    palette='colorblind',
    order=df['cuisine'].value_counts().index # Orders bars from most to least frequent
)

# 4. Add titles and labels
plt.title('Zomato: Distribution of Food Orders by Cuisine Type', fontsize=14, pad=15)
plt.xlabel('Cuisine Type', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

# 5. Display the plot
plt.show()
```

Generate a Seaborn lineplot of daily step counts (simulate data for 2 weeks) and style it using the 'muted' palette and 'notebook' context for a clean, presentation-ready look.

```python
np.random.seed(42)

# 1. Create simulated step count data for 2 weeks (14 days)
dates = pd.date_range(start='2026-10-01', periods=14, freq='D')
steps = np.random.randint(5000, 14000, size=14)

df = pd.DataFrame({
    'Date': dates,
    'Steps': steps
})

# 2. Apply the 'notebook' context, whitegrid style, and 'muted' palette
sns.set_theme(context="notebook", style="whitegrid", palette="muted")

# 3. Create the lineplot
plt.figure(figsize=(10, 5))
sns.lineplot(
    data=df,
    x='Date',
    y='Steps',
    marker='o',
    linewidth=2.5,
    markersize=8
)

# 4. Add titles, axis labels, and format tick labels
plt.title('Daily Step Counts Over Two Weeks', fontsize=14, pad=15)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Step Count', fontsize=12)
plt.xticks(rotation=30)
plt.tight_layout()

# 5. Display the plot
plt.show()
```

Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use sns.set() to set a 'deep' palette, 'talk' context, and customize the font scale and grid style. Apply your theme to a barplot showing sales of 5 product categories.

```python
sns.set_theme(
    context="talk",
    style="whitegrid",
    palette="deep",
    font_scale=1.1
)

# 2. Create a sample Flipkart sales dataset (5 product categories)
categories = ['Mobiles', 'Electronics', 'Fashion', 'Groceries', 'Home Appliances']
sales_in_crores = [45.5, 32.0, 28.5, 19.0, 24.2]

df = pd.DataFrame({
    'Category': categories,
    'Sales': sales_in_crores
})

# 3. Create the barplot
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df,
    x='Category',
    y='Sales',
    edgecolor='black',
    linewidth=0.8
)

# 4. Add titles and axis labels
plt.title('Flipkart Sales Dashboard: Q3 Revenue by Category', pad=20, fontweight='bold')
plt.xlabel('Product Category')
plt.ylabel('Sales (in ₹ Crores)')

# 5. Display the plot
plt.show()
```

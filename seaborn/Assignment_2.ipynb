```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

Use sns.countplot() to visualize the number of orders placed by each payment method (e.g., UPI, Card, Cash) in a sample Swiggy orders DataFrame with at least 3 payment types.

```python
data = {
    'Order_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    'Payment_Method': [
        'UPI', 'Credit Card', 'Cash on Delivery', 'UPI', 'UPI', 
        'Credit Card', 'Cash on Delivery', 'UPI', 'Credit Card', 
        'UPI', 'Cash on Delivery', 'UPI'
    ]
}
df = pd.DataFrame(data)

sns.set_theme(style="whitegrid")

sns.countplot(data=df, x='Payment_Method',hue='Payment_Method', palette='Set2')
plt.title("Number of Orders by Payment Method")
plt.xlabel("Payment_Method")
plt.show()
```

Create a barplot using sns.barplot() to show the average delivery time (in minutes) for each food category (e.g., Pizza, Burger, Biryani, South Indian) from a Zomato-style dataset.<br><br><em><strong>Hint:</strong> Use the estimator parameter to show the mean delivery time for each category.</em>

```python
data = {
    'Food_Category': [
        'Pizza', 'Burger', 'Biryani', 'South Indian', 
        'Pizza', 'Burger', 'Biryani', 'South Indian', 
        'Pizza', 'Burger', 'Biryani', 'South Indian'
    ],
    'Delivery_Time_Minutes': [
        35, 25, 45, 30, 
        40, 20, 50, 25, 
        30, 28, 42, 35
    ]
}
df = pd.DataFrame(data)

sns.set_theme(style="whitegrid")
sns.barplot(data=df, x='Food_Category', y='Delivery_Time_Minutes',hue='Food_Category', estimator='mean', palette='Blues_d')
plt.show()
```

Modify your barplot to display the total number of orders instead of the average by setting estimator to sum in sns.barplot().

```python
data = {
    'Food_Category': [
        'Pizza', 'Burger', 'Biryani', 'South Indian', 
        'Pizza', 'Burger', 'Biryani', 'South Indian', 
        'Pizza', 'Burger', 'Biryani', 'South Indian'
    ],
    'Orders': [
        15, 20, 25, 10, 
        12, 18, 30, 14, 
        18, 22, 28, 12
    ]
}
df = pd.DataFrame(data)

sns.set_theme(style="whitegrid")
sns.barplot(data=df, x='Food_Category',y='Orders', hue='Food_Category', estimator='sum',palette='viridis')
plt.show()
```

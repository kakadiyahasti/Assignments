Create a Pandas DataFrame with at least 5 columns representing different features of food delivery orders (e.g., delivery_time, order_value, tip_amount, rating, delivery_distance) and 20 rows of random data, then calculate and display the correlation matrix using the .corr() function.

```python
np.random.seed(42)

# Number of rows
n = 20

# Generate random data for 5 food delivery features
data = {
    'delivery_time': np.random.randint(15, 60, size=n),  # Minutes
    'order_value': np.random.randint(200, 1500, size=n),  # ₹
    'tip_amount': np.random.randint(10, 200, size=n),  # ₹
    'rating': np.round(np.random.uniform(3.0, 5.0, size=n), 1),  # Stars
    'delivery_distance': np.round(
        np.random.uniform(1.0, 12.0, size=n), 1
    ),  # Kilometers
}

# Create the DataFrame
df = pd.DataFrame(data)

# Calculate the correlation matrix
correlation_matrix = df.corr()

# Display the DataFrame and its correlation matrix
print('--- Food Delivery Orders DataFrame ---')
print(df)

print('\n--- Correlation Matrix ---')
print(correlation_matrix)
```

Using Seaborn, plot a heatmap of the correlation matrix you generated for the food delivery data, and add annotations to show the correlation values inside each cell.

```python
np.random.seed(42)

# Number of rows
n = 20

# Generate random data for 5 food delivery features
data = {
    'delivery_time': np.random.randint(15, 60, size=n),  # Minutes
    'order_value': np.random.randint(200, 1500, size=n),  # ₹
    'tip_amount': np.random.randint(10, 200, size=n),  # ₹
    'rating': np.round(np.random.uniform(3.0, 5.0, size=n), 1),  # Stars
    'delivery_distance': np.round(
        np.random.uniform(1.0, 12.0, size=n), 1
    ),  # Kilometers
}

# Create the DataFrame
df = pd.DataFrame(data)

# Calculate the correlation matrix
correlation_matrix = df.corr()

# Set up the matplotlib figure
plt.figure(figsize=(8, 6))

# Plot the heatmap with annotations
sns.heatmap(
    correlation_matrix,
    annot=True,  # Displays the correlation values inside each cell
    cmap='coolwarm',  # Color palette (blue for negative, red for positive)
    fmt='.2f',  # Formats correlation values to 2 decimal places
    linewidths=0.5,  # Adds grid lines between cells
    vmin=-1,
    vmax=1,  # Fixes the color scale range from -1 to 1
    cbar=True,
)

# Add title
plt.title(
    'Correlation Matrix Heatmap of Food Delivery Features',
    fontsize=13,
    fontweight='bold',
    pad=15,
)

plt.tight_layout()
plt.show()
```

Change the color palette of your Seaborn heatmap to a diverging palette (such as 'coolwarm' or 'RdYlGn') to better visualize positive and negative correlations, and explain which two features in your data have the strongest negative correlation.<br><br><em><strong>Hint:</strong> Use the cmap parameter in sns.heatmap().</em>

```python
np.random.seed(42)

# Number of rows
n = 20

# Generate random data for 5 food delivery features
data = {
    'delivery_time': np.random.randint(15, 60, size=n),  # Minutes
    'order_value': np.random.randint(200, 1500, size=n),  # ₹
    'tip_amount': np.random.randint(10, 200, size=n),  # ₹
    'rating': np.round(np.random.uniform(3.0, 5.0, size=n), 1),  # Stars
    'delivery_distance': np.round(
        np.random.uniform(1.0, 12.0, size=n), 1
    ),  # Kilometers
}

# Create the DataFrame
df = pd.DataFrame(data)

# Calculate the correlation matrix
correlation_matrix = df.corr()

# Programmatically find the strongest negative correlation (excluding self-correlation of 1.0)
corr_unstacked = (
    correlation_matrix.where(~np.eye(correlation_matrix.shape[0], dtype=bool))
    .unstack()
    .dropna()
)
strongest_neg_pair = corr_unstacked.idxmin()
strongest_neg_val = corr_unstacked.min()

print(
    f'Strongest negative correlation: "{strongest_neg_pair[0]}" and "{strongest_neg_pair[1]}" (Coefficient: {strongest_neg_val:.2f})'
)

# Set up the matplotlib figure
plt.figure(figsize=(8, 6))

# Plot the heatmap with a diverging palette ('RdYlGn' or 'coolwarm') and annotations
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='RdYlGn',  # Diverging palette: Red for negative, Yellow for neutral, Green for positive
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    cbar=True,
)

# Add title
plt.title(
    'Correlation Matrix Heatmap (Diverging Palette: RdYlGn)',
    fontsize=13,
    fontweight='bold',
    pad=15,
)

plt.tight_layout()
plt.show()
```

Take the Flipkart mobile phones dataset (or any small CSV with product features you can find online), compute the correlation matrix for at least 4 numeric columns (e.g., price, rating, RAM, battery_capacity), and generate a Seaborn heatmap with annotations and a diverging color palette.

```python
np.random.seed(42)

# Number of mobile phone samples
n = 30

# Generate sample data representing Flipkart mobile features
data = {
    'price': np.random.randint(8000, 75000, size=n),              # Price in ₹
    'rating': np.round(np.random.uniform(3.2, 4.8, size=n), 1),    # Star Rating
    'ram_gb': np.random.choice([4, 6, 8, 12, 16], size=n),         # RAM in GB
    'battery_mah': np.random.choice([4000, 4500, 5000, 6000], size=n) # Battery in mAh
}

# Create the DataFrame
df = pd.DataFrame(data)

# Compute the correlation matrix
correlation_matrix = df.corr()

# Set up the matplotlib figure
plt.figure(figsize=(8, 6))

# Generate the heatmap with annotations and a diverging palette ('coolwarm')
sns.heatmap(
    correlation_matrix,
    annot=True,          # Displays correlation values inside each cell
    cmap='coolwarm',     # Diverging color palette (blue for negative, red for positive)
    fmt='.2f',           # Formats numbers to 2 decimal places
    linewidths=0.5,      # Adds grid lines between cells
    vmin=-1, 
    vmax=1               # Fixes scale from -1 to 1
)

# Add title and layout adjustments
plt.title('Correlation Matrix Heatmap: Flipkart Mobile Phones', fontsize=13, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```

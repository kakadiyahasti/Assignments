```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Create a Seaborn barplot showing the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and save the chart as a PNG file with 300 DPI.

```python
data = {
    'Food Category': [
        'Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'
    ],
    'Number of Orders': [
        150, 120, 180, 90, 110
    ]
}

df = pd.DataFrame(data)

# Set plot style
sns.set_theme(style="whitegrid")

# Create barplot
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=df,
    x='Food Category',
    y='Number of Orders',
    hue='Food Category',
    palette='viridis',
    legend=False
)

# Add labels and title
plt.title('Number of Orders per Food Category',
          fontsize=16, fontweight='bold')
plt.xlabel('Food Category', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

# Display values above bars
for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=10)

plt.tight_layout()

# Save chart as PNG at 300 DPI
plt.savefig(
    'food_category_orders.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

Generate a Seaborn line plot visualizing daily step counts for one week, then export the chart as an SVG file suitable for use in a PowerPoint presentation.<br><br><em><strong>Hint:</strong> Use plt.savefig('filename.svg', format='svg').</em>

```python
data = {
    'Day': [
        'Monday', 'Tuesday', 'Wednesday',
        'Thursday', 'Friday', 'Saturday', 'Sunday'
    ],
    'Steps': [
        4500, 6200, 7800, 5500, 9000, 11000, 8500
    ]
}

df = pd.DataFrame(data)

# Set Seaborn style
sns.set_theme(style="whitegrid")

# Create figure
plt.figure(figsize=(10, 6))

# Create line plot
sns.lineplot(
    data=df,
    x='Day',
    y='Steps',
    marker='o',
    linewidth=2.5,
    markersize=9,
    color='royalblue'
)

# Add title and labels
plt.title(
    'Daily Step Counts Over One Week',
    fontsize=16,
    fontweight='bold'
)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Number of Steps', fontsize=12)

# Rotate x-axis labels if needed
plt.xticks(rotation=30)

# Improve layout
plt.tight_layout()

# Export as SVG for PowerPoint
plt.savefig(
    'weekly_step_counts.svg',
    format='svg',
    bbox_inches='tight'
)

# Display plot
plt.show()
```

Customize a Seaborn chart using the 'darkgrid' style and a color palette of your choice, then save it as a high-resolution PNG image with 1200 DPI

```python
data = {
    'Food Category': [
        'Pizza', 'Burger', 'Biryani',
        'Momos', 'Ice Cream'
    ],
    'Orders': [150, 120, 180, 90, 110]
}

df = pd.DataFrame(data)

# Set Seaborn style and color palette
sns.set_style('darkgrid')
sns.set_palette('magma')

# Create figure
plt.figure(figsize=(10, 6))

# Create customized bar chart
ax = sns.barplot(
    data=df,
    x='Food Category',
    y='Orders',
    hue='Food Category',
    palette='magma',
    legend=False
)

# Customize title and axis labels
plt.title(
    'Food Category vs Number of Orders',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Food Category', fontsize=13)
plt.ylabel('Number of Orders', fontsize=13)

# Add values above bars
for container in ax.containers:
    ax.bar_label(container, padding=5, fontsize=11)

# Adjust layout
plt.tight_layout()

# Save high-resolution PNG at 1200 DPI
plt.savefig(
    'customized_seaborn_chart.png',
    format='png',
    dpi=1200,
    bbox_inches='tight'
)

# Display chart
plt.show()
```

Create a template function called apply_presentation_theme() that applies a consistent Seaborn style, font, and color palette to any chart, then use it to style and export a pie chart showing your top 5 most-used apps.

```python

def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        context="talk",
        palette="Set2",
        font="Arial"
    )

    plt.rcParams.update({
        "figure.figsize": (10, 7),
        "figure.dpi": 120,
        "font.family": "Arial",
        "font.size": 12,
        "axes.titlesize": 20,
        "axes.titleweight": "bold",
        "axes.labelsize": 14,
        "savefig.bbox": "tight"
    })


# 2. Apply presentation theme
apply_presentation_theme()

# 3. App usage data (illustrative)
apps = ["WhatsApp", "Instagram", "YouTube", "Chrome", "Spotify"]
usage = [30, 25, 20, 15, 10]

# 4. Create pie chart
colors = sns.color_palette("Set2", n_colors=5)

fig, ax = plt.subplots(figsize=(10, 7))

ax.pie(
    usage,
    labels=apps,
    colors=colors,
    autopct="%1.1f%%",
    startangle=140,
    shadow=False,
    textprops={"fontsize": 12},
    wedgeprops={"edgecolor": "white", "linewidth": 2}
)

# 5. Add chart title
ax.set_title(
    "Top 5 Most-Used Apps",
    pad=20
)

ax.axis("equal")

# 6. Export chart as high-resolution PNG
plt.savefig(
    "top_5_most_used_apps.png",
    dpi=300,
    format="png",
    bbox_inches="tight"
)

# 7. Export as SVG for presentations
plt.savefig(
    "top_5_most_used_apps.svg",
    format="svg",
    bbox_inches="tight"
)

# 8. Display chart
plt.show()
```

Use ChatGPT to generate a code snippet for exporting a Seaborn heatmap as a 1200 DPI SVG file, then test and adapt the code to save your own chart.<br><br><em><strong>Hint:</strong> Paste your prompt and the code you received as a comment in your script.</em>

```python

# =====================================================
# ChatGPT Prompt:
# "Generate a code snippet for exporting a Seaborn
# heatmap as a 1200 DPI SVG file, then test and adapt
# the code to save your own chart."
#
# ChatGPT Generated Code:
# plt.savefig('heatmap.svg', format='svg', dpi=1200)
# =====================================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. Create sample data
data = np.array([
    [85, 70, 90, 65, 80],
    [75, 88, 60, 95, 72],
    [90, 78, 85, 70, 92],
    [65, 80, 75, 88, 60],
    [95, 85, 70, 80, 90]
])

categories = ['Math', 'Science', 'English', 'History', 'Computer']

df = pd.DataFrame(
    data,
    index=['Student 1', 'Student 2', 'Student 3',
           'Student 4', 'Student 5'],
    columns=categories
)

# 2. Apply Seaborn theme
sns.set_theme(style="white", font_scale=1.1)

# 3. Create heatmap
plt.figure(figsize=(10, 7))

ax = sns.heatmap(
    df,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=1,
    linecolor="white",
    cbar_kws={'label': 'Marks'}
)

# 4. Customize chart
plt.title(
    'Student Performance Heatmap',
    fontsize=18,
    fontweight='bold',
    pad=15
)

plt.xlabel('Subjects', fontsize=12)
plt.ylabel('Students', fontsize=12)

plt.tight_layout()

# 5. Export as SVG at 1200 DPI
plt.savefig(
    'student_performance_heatmap.svg',
    format='svg',
    dpi=1200,
    bbox_inches='tight'
)

# 6. Display chart
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.

```python
np.random.seed(42)
delivery_times = np.random.randint(15, 65, size=35)
df = pd.DataFrame({'Delivery_Time': delivery_times})


sns.histplot(data=df, bins=10, kde=True, color='coral')
plt.show()
```

Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the distribution using sns.kdeplot(), showing the probability density curve.

```python
np.random.seed(42)
steps = np.random.normal(loc=8500, scale=1800, size=30).astype(int)
steps = np.clip(steps, 3000, 15000)

df = pd.DataFrame({'Daily_Steps': steps})

plt.figure(figsize=(9, 5))
sns.kdeplot(
    data=df, 
    x='Daily_Steps', 
    fill=True,           
    color='rebeccapurple', 
    linewidth=2.5,
    bw_adjust=0.8        
)
plt.show()
```

Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:</strong> Use the same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>

```python
np.random.seed(42)

spending = np.random.normal(loc=3500, scale=1000, size=40).astype(int)
spending = np.clip(spending, 1200, 7500) 
df = pd.DataFrame({'Monthly_Spending': spending})

sns.histplot(
    data=df, 
    x='Monthly_Spending', 
    bins=8, 
    stat='density', 
    color='skyblue', 
    edgecolor='black', 
    alpha=0.6
)

sns.kdeplot(
    data=df, 
    x='Monthly_Spending', 
    color='crimson', 
    linewidth=2.5, 
    fill=True, 
    alpha=0.2
)

plt.show()
```

Given a dataset of IPL match scores, use sns.histplot() to plot the run distribution. Change the number of bins and observe how the plot changes. Write 2-3 lines about what you notice.<br><br><em><strong>Hint:</strong> Try bins=5, bins=10, and bins=20.</em>

```python
scores = [111,120,145,160,175,180,190,155,200,210,165,140,185,195,120,170,205,230,175,160,190,215,145,180,200,240,155,170,185,225]

df = pd.DataFrame(scores, columns=['scores'])

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(data=df, x='scores', bins=5, ax=axes[0], color='skyblue')
axes[0].set_title('bins = 5')

sns.histplot(data=df, x='scores', bins=10, ax=axes[1], color='salmon')
axes[1].set_title('bins = 10')

sns.histplot(data=df, x='scores', bins=20, ax=axes[2], color='lightgreen')
axes[2].set_title('bins = 20')

plt.tight_layout()
plt.show()
```

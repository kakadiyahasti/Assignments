```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Use sns.scatterplot() to visualize the relationship between the number of followers and the number of posts for 20 fictional Instagram users. Label the axes appropriately.

```python
np.random.seed(42)

# Number of fictional Instagram users
n = 20

# Generate random data for posts (10 to 500) and followers (100 to 50,000)
posts = np.random.randint(10, 500, size=n)
followers = np.random.randint(100, 50000, size=n)

# Create a DataFrame
df = pd.DataFrame({
    'posts': posts,
    'followers': followers
})

# Set up the matplotlib figure
plt.figure(figsize=(9, 5))

# Create the scatter plot using Seaborn
sns.scatterplot(
    data=df,
    x='posts',
    y='followers',
    color='purple',
    s=100,      # Marker size
    alpha=0.8   # Transparency
)

# Add titles and appropriately label the axes
plt.title('Instagram Users: Number of Posts vs. Followers', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Posts', fontsize=12)
plt.ylabel('Number of Followers', fontsize=12)

plt.tight_layout()
plt.show()
```

Create a line plot using sns.lineplot() to show the daily number of food orders on Zomato for each day over a month. Use a list of 30 random order counts to simulate the data.

```python
np.random.seed(42)

# Generate 30 days of simulated Zomato order counts (e.g., fluctuating between 1,500 and 4,500 orders)
days = np.arange(1, 31)
order_counts = np.random.randint(1500, 4500, size=30)

# Create a DataFrame
df = pd.DataFrame({
    'day': days,
    'orders': order_counts
})

# Set up the matplotlib figure
plt.figure(figsize=(10, 5))

# Create the Line Plot using Seaborn
sns.lineplot(
    data=df,
    x='day',
    y='orders',
    marker='o',          # Adds markers at each data point
    color='orangered',   # Zomato-themed color
    linewidth=2.5
)

# Add titles and appropriately label the axes
plt.title('Zomato Daily Food Orders Over a Month', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Day of the Month', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

# Set x-axis ticks to show every day clearly
plt.xticks(np.arange(1, 31, 2))

plt.tight_layout()
plt.show()
```

Using sns.relplot(), plot the relationship between movie ratings and the number of reviews for different genres (Action, Comedy, Drama) based on a sample dataset of 15 movies. Use the 'hue' parameter to differentiate genres.<br><br><em><strong>Hint:</strong> Create a DataFrame with columns: 'movie', 'rating', 'reviews', and 'genre'.</em>

```python
np.random.seed(42)

# Sample dataset of 15 movies across different features
movies = [
    "Movie A", "Movie B", "Movie C", "Movie D", "Movie E",
    "Movie F", "Movie G", "Movie H", "Movie I", "Movie J",
    "Movie K", "Movie L", "Movie M", "Movie N", "Movie O"
]
genres = np.random.choice(['Action', 'Comedy', 'Drama'], size=15)
ratings = np.round(np.random.uniform(5.0, 9.5, size=15), 1)
reviews = np.random.randint(500, 50000, size=15)

# Create the DataFrame with the requested columns
df = pd.DataFrame({
    'movie': movies,
    'rating': ratings,
    'reviews': reviews,
    'genre': genres
})

# Create the Relational Plot using sns.relplot()
g = sns.relplot(
    data=df,
    x='reviews',
    y='rating',
    hue='genre',
    kind='scatter',
    s=120,          # Marker size
    palette='Set1', # Distinct color palette for genres
    height=5,
    aspect=1.4
)

# Add titles and axis labels
g.fig.suptitle('Movie Ratings vs. Number of Reviews by Genre', fontsize=14, fontweight='bold', y=1.03)
g.set_axis_labels('Number of Reviews', 'Rating (out of 10)', fontsize=12)

plt.show()
```

Generate a regression plot using sns.regplot() to analyze the correlation between the number of hours spent on YouTube and exam scores for a group of 25 students. Interpret whether the trend is positive or negative.

```python
np.random.seed(42)

# Number of students
n = 25

# Generate hours spent on YouTube (between 0.5 and 6 hours)
youtube_hours = np.random.uniform(0.5, 6.0, size=n)

# Generate exam scores inversely related to YouTube hours (with random noise)
exam_scores = (95 - (youtube_hours * 9) + np.random.normal(0, 5, size=n)).clip(40, 100)

# Create a DataFrame
df = pd.DataFrame({
    'youtube_hours': youtube_hours,
    'exam_score': exam_scores
})

# Set up the matplotlib figure
plt.figure(figsize=(9, 5))

# Create the Regression Plot using Seaborn
sns.regplot(
    data=df,
    x='youtube_hours',
    y='exam_score',
    color='darkred',
    marker='o'
)

# Add titles and labels
plt.title('Impact of Daily YouTube Hours on Exam Scores', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Hours Spent on YouTube (Daily)', fontsize=12)
plt.ylabel('Exam Score (%)', fontsize=12)

plt.tight_layout()
plt.show()
```

Using Seaborn's lineplot(), plot the daily temperature trend for one week using the following data: days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'], temps = [32, 34, 33, 35, 36, 37, 34]. Label the axes and title the chart 'Weekly Temperature Trend'.

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps)
plt.title("weekly temprature trend")
plt.xlabel("Days")
plt.ylabel("temprature")
plt.show()
```

Recreate the above line plot using Seaborn's axes-level and figure-level functions separately. Briefly note one difference you observe in how these functions work.<br><br><em><strong>Hint:</strong> Use sns.lineplot() for axes-level and sns.relplot() for figure-level.</em>

```python
data = pd.DataFrame({
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'Temp': [32, 34, 33, 35, 36, 37, 34]
})


sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(8, 5))
sns.lineplot(data=data, x='Day', y='Temp', marker='o', color='teal', ax=ax, linewidth=2.5, markersize=8)

ax.set_title('Weekly Temperature Trend', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Day', fontsize=12)
ax.set_ylabel('Temperature (°C)', fontsize=12)
plt.show()

g = sns.relplot(
    data=data, 
    x='Day', 
    y='Temp', 
    kind='line', 
    marker='o', 
    color='teal', 
    height=5, 
    aspect=1.6
)

g.fig.suptitle('Weekly Temperature Trend', fontsize=14, fontweight='bold', y=1.03)
g.set_axis_labels('Day', 'Temperature (°C)')

plt.show()
```

Apply each of Seaborn's built-in themes (darkgrid, whitegrid, ticks) to your temperature trend plot and save three separate images. Compare how the grid and background change with each theme.<br><br><em><strong>Hint:</strong> Use sns.set_theme(style='theme_name') before plotting.</em>

```python
themes = ['darkgrid', 'whitegrid', 'ticks']

for theme in themes:
    
    sns.set_theme(style=theme)
    
    plt.figure(figsize=(8, 5))
    
    
    sns.lineplot(data=data, x='Day', y='Temp', marker='o', color='teal', linewidth=2.5, markersize=8)
    
    
    plt.title(f'Weekly Temperature Trend ({theme.capitalize()} Theme)', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Day', fontsize=12)
    plt.ylabel('Temperature (°C)', fontsize=12)
    
    
    filename = f'temperature_trend_{theme}.png'
    plt.show()
    # plt.savefig(filename, dpi=300, bbox_inches='tight')
    # print(f"Successfully saved: {filename}")
    
    
    plt.close()
```

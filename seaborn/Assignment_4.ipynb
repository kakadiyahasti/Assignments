```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution.

```python
prices = [299, 499, 599, 799, 999, 1299, 1499, 1799, 1999, 2499, 
          2999, 3499, 3999, 4499, 4999, 5999, 6999, 7999, 9999, 14999]

# Convert the list to a DataFrame with a column name
df = pd.DataFrame(prices, columns=['price'])

# Set up the plot figure
plt.figure(figsize=(10, 5))

# Create the Box Plot using Seaborn
sns.boxplot(data=df, x='price', color='lightcoral', showfliers=True)

# Add titles and labels
plt.title('Distribution of Flipkart Product Prices', fontsize=14, fontweight='bold')
plt.xlabel('Price (₹)', fontsize=12)

plt.tight_layout()
plt.show()
```

Use sns.violinplot() to compare the order amounts for two different Zomato restaurant categories (e.g., 'Fast Food' vs 'Desserts') from a sample dataset.

```python
np.random.seed(42)

n = 100
fast_food_amounts = np.random.normal(loc=350, scale=100, size=n).clip(150, 800)
dessert_amounts = np.random.normal(loc=250, scale=75, size=n).clip(100, 600)

df = pd.DataFrame({
    'category': ['Fast Food'] * n + ['Desserts'] * n,
    'order_amount': list(fast_food_amounts) + list(dessert_amounts)
})

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x='category',
    y='order_amount',
    palette=['skyblue', 'lightpink'],
    inner='quartile'  
)

plt.title('Comparison of Order Amounts: Fast Food vs Desserts', fontsize=14, fontweight='bold')
plt.xlabel('Restaurant Category', fontsize=12)
plt.ylabel('Order Amount (₹)', fontsize=12)

plt.tight_layout()
plt.show()
```

Given an array of IPL match ticket prices, plot a boxplot using sns.boxplot() and identify any outliers in the data.<br><br><em><strong>Hint:</strong> Outliers are points that appear outside the whiskers of the boxplot.</em>

```python
ticket_prices = [800, 1000, 1200, 1400, 1500, 1600, 1800, 2000, 
                 2200, 2500, 2800, 3000, 3500, 4000, 9500, 12000]

df = pd.DataFrame(ticket_prices, columns=['price'])

Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]['price'].tolist()
print(f"Identified Outliers: {outliers}")

plt.figure(figsize=(9, 4))

sns.boxplot(data=df, x='price', color='lightblue', showfliers=True)

plt.title('IPL Match Ticket Prices Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Price (₹)', fontsize=12)

plt.tight_layout()
plt.show()
```

Create a violin plot using sns.violinplot() to show the distribution of daily step counts collected from a fitness app like Google Fit, and explain what the shape of the plot tells you about user activity.

```python
np.random.seed(42)

n = 150
weekday_steps = np.random.normal(loc=8500, scale=2200, size=n).clip(2000, 16000)
weekend_steps = np.random.normal(loc=6800, scale=3000, size=n).clip(1000, 15000)

df = pd.DataFrame({
    'day_type': ['Weekday'] * n + ['Weekend'] * n,
    'step_count': list(weekday_steps) + list(weekend_steps)
})

plt.figure(figsize=(9, 5))

sns.violinplot(
    data=df,
    x='day_type',
    y='step_count',
    palette=['skyblue', 'lightgreen'],
    inner='quartile'  )


plt.title('Daily Step Count Distribution (Google Fit Simulation)', fontsize=14, fontweight='bold')
plt.xlabel('Day Type', fontsize=12)
plt.ylabel('Steps', fontsize=12)

plt.tight_layout()
plt.show()
```

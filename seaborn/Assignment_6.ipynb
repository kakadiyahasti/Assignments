```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Load a CSV file containing Flipkart product ratings and prices, then use Seaborn's jointplot with kind='scatter' to visualize the relationship between rating and price.

```python
np.random.seed(42)

n = 300
ratings = np.random.uniform(1.0, 5.0, size=n).round(1)
prices = np.random.exponential(scale=3000, size=n).clip(100, 50000)

df = pd.DataFrame({
    'rating': ratings,
    'price': prices
})

sns.jointplot(
    data=df,
    x='rating',
    y='price',
    kind='scatter',
    color='royalblue',
    height=7,
    ratio=5,
    alpha=0.6
)

plt.show()
```

Create a jointplot with kind='hex' to analyze the relationship between delivery time and order value using a dataset of Swiggy food orders.<br><br><em><strong>Hint:</strong> Use sns.jointplot(x='delivery_time', y='order_value', data=your_dataframe, kind='hex').</em>

```python
np.random.seed(42)

n = 1000
delivery_times = np.random.normal(loc=35, scale=10, size=n).clip(15, 90)
order_values = np.random.normal(loc=450, scale=200, size=n).clip(150, 2500)

df = pd.DataFrame({
    'delivery_time': delivery_times,
    'order_value': order_values
})

sns.jointplot(
    data=df,
    x='delivery_time',
    y='order_value',
    kind='hex',
    color='darkorange',
    height=7
)

plt.show()
```

Use Seaborn's jointplot with kind='kde' to visualize how the number of likes and comments are related in a dataset of Instagram posts.

```python
np.random.seed(42)

n = 500
likes = np.random.exponential(scale=2000, size=n).astype(int) + 100

comments = (likes * np.random.uniform(0.01, 0.08, size=n)).astype(int) + 5

df = pd.DataFrame({
    'likes': likes,
    'comments': comments
})

sns.jointplot(
    data=df,
    x='likes',
    y='comments',
    kind='kde',
    cmap='Purples',
    fill=True,
    height=7
)

plt.show()
```

Compare the visual insights you get from scatter, hexbin, and KDE jointplots using the same dataset of Zomato restaurant reviews (rating vs number of reviews). Briefly describe which plot type helps you understand the data best and why.

```python
np.random.seed(42)

# Generate mock Zomato dataset: rating (1.0 to 5.0) and right-skewed review counts
n = 1000
ratings = np.random.uniform(1.0, 5.0, size=n).round(1)
reviews = np.random.exponential(scale=300, size=n).astype(int) + 5

# Create a DataFrame
df = pd.DataFrame({
    'rating': ratings,
    'reviews': reviews
})

# --- 1. Scatter Jointplot ---
sns.jointplot(
    data=df,
    x='rating',
    y='reviews',
    kind='scatter',
    color='royalblue',
    alpha=0.5,
    height=6
)
plt.suptitle('Zomato: Scatter Jointplot', y=1.02, fontsize=12, fontweight='bold')
plt.show()

# --- 2. Hexbin Jointplot ---
sns.jointplot(
    data=df,
    x='rating',
    y='reviews',
    kind='hex',
    color='darkorange',
    height=6
)
plt.suptitle('Zomato: Hexbin Jointplot', y=1.02, fontsize=12, fontweight='bold')
plt.show()

# --- 3. KDE Jointplot ---
sns.jointplot(
    data=df,
    x='rating',
    y='reviews',
    kind='kde',
    cmap='Purples',
    fill=True,
    height=6
)
plt.suptitle('Zomato: KDE Jointplot', y=1.02, fontsize=12, fontweight='bold')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Use Seaborn's regplot to visualize the relationship between the number of likes and number of comments on a set of Instagram posts (create a small DataFrame with at least 10 sample data points).

```python
data = {
    'likes': [120, 450, 85, 1250, 310, 620, 95, 2300, 540, 780, 45, 1600],
    'comments': [12, 45, 8, 120, 25, 60, 10, 210, 50, 75, 4, 150]
}
df = pd.DataFrame(data)

# 2. Set the visual style
sns.set_theme(style="whitegrid")

# 3. Create the regplot
plt.figure(figsize=(8, 6))
sns.regplot(
    data=df,
    x='likes',
    y='comments',
    color='purple',
    marker='o',
    line_kws={'color': 'darkorange'} # Customize regression line color
)

# 4. Add titles and labels
plt.title('Relationship Between Likes and Comments on Instagram Posts', fontsize=14, pad=15)
plt.xlabel('Number of Likes', fontsize=12)
plt.ylabel('Number of Comments', fontsize=12)

# 5. Display the plot
plt.show()
```

Create a lmplot in Seaborn to show the regression line between food order price and delivery time using a sample Zomato-style dataset (minimum 15 rows with random values for price and delivery time).

```python
np.random.seed(42)

# 1. Create a sample Zomato-style DataFrame with 20 rows
n_rows = 20
data = {
    'price': np.random.randint(150, 1500, size=n_rows),
    'delivery_time': np.random.randint(20, 75, size=n_rows),
    'cuisine': np.random.choice(['North Indian', 'Chinese', 'Fast Food', 'South Indian'], size=n_rows)
}
df = pd.DataFrame(data)

# 2. Set the visual style
sns.set_theme(style="whitegrid")

# 3. Create the lmplot with hue differentiation by cuisine
plot = sns.lmplot(
    data=df,
    x='price',
    y='delivery_time',
    hue='cuisine',
    height=6,
    aspect=1.3,
    palette='Set2',
    markers=['o', 's', '^', 'D']
)

# 4. Add titles and labels
plot.set_axis_labels('Order Price (INR)', 'Delivery Time (Minutes)')
plot.fig.suptitle('Food Order Price vs. Delivery Time by Cuisine', y=1.03, fontsize=14)

# 5. Display the plot
plt.show()
```

Apply lowess smoothing to a regplot that visualizes the relationship between the number of songs in a playlist and total playlist duration (in minutes) for Spotify users.<br><br><em><strong>Hint:</strong> Use the lowess=True parameter in regplot to see the difference in fit.</em>

```python
np.random.seed(42)

# 1. Create a sample Spotify playlist DataFrame with 18 rows
n_rows = 18
songs = np.random.randint(10, 250, size=n_rows)
# Simulate duration roughly proportional to songs (avg ~3.5 mins per song) with noise
duration_minutes = songs * 3.5 + np.random.normal(0, 15, size=n_rows)
duration_minutes = np.clip(duration_minutes, 15, 900) # Ensure realistic values

df = pd.DataFrame({
    'songs': songs,
    'duration_minutes': duration_minutes
})

# 2. Set the visual style
sns.set_theme(style="whitegrid")

# 3. Create the regplot with LOWESS smoothing
plt.figure(figsize=(8, 6))
sns.regplot(
    data=df,
    x='songs',
    y='duration_minutes',
    # lowess=True,  # Enables non-parametric LOWESS smoothing fit
    color='forestgreen',
    marker='o',
    line_kws={'color': 'darkred', 'linewidth': 2}
)

# 4. Add titles and labels
plt.title('Spotify Playlist: Number of Songs vs. Total Duration (LOWESS Fit)', fontsize=14, pad=15)
plt.xlabel('Number of Songs', fontsize=12)
plt.ylabel('Total Duration (Minutes)', fontsize=12)

# 5. Display the plot
plt.show()
```

Generate a faceted regression plot (using lmplot with col parameter) to compare the relationship between hours spent on YouTube and exam scores, split by 'preparation method' (e.g., 'self-study' vs 'coaching') in your DataFrame.

```python
np.random.seed(42)

# 1. Create a sample DataFrame with 25 rows
n_rows = 25
hours_youtube = np.random.uniform(0.5, 6.0, size=n_rows)
preparation_method = np.random.choice(['Self-Study', 'Coaching'], size=n_rows)

# Simulate exam scores: higher YouTube hours slightly lower scores, with coaching having a higher baseline
base_score = np.where(preparation_method == 'Coaching', 85, 75)
exam_score = base_score - (hours_youtube * 3.5) + np.random.normal(0, 4, size=n_rows)
exam_score = np.clip(exam_score, 40, 100) # Keep scores within realistic bounds

df = pd.DataFrame({
    'hours_youtube': hours_youtube,
    'exam_score': exam_score,
    'preparation_method': preparation_method
})

# 2. Set the visual style
sns.set_theme(style="whitegrid")

# 3. Create the faceted lmplot using the 'col' parameter
plot = sns.lmplot(
    data=df,
    x='hours_youtube',
    y='exam_score',
    col='preparation_method', # Facets the plot into separate panels columns
    hue='preparation_method',  # Colors the points and lines by method
    height=5,
    aspect=1.1,
    palette='muted',
    markers=['o', 's']
)

# 4. Add titles and axis labels
plot.set_axis_labels('Hours Spent on YouTube', 'Exam Score')
plot.fig.suptitle('Impact of YouTube Hours on Exam Scores by Preparation Method', y=1.03, fontsize=14)

# 5. Display the plot
plt.show()
```

Modify a given lmplot that shows regression between Flipkart product price and number of reviews. Add a title, axis labels, and customize the marker color.<br><br><em><strong>Constraint:</strong> Do not change the regression logic—only update plot aesthetics.</em>

```python
np.random.seed(42)

# 1. Create a sample Flipkart product dataset
n_rows = 20
df = pd.DataFrame({
    'price': np.random.randint(200, 5000, size=n_rows),
    'reviews': np.random.randint(10, 800, size=n_rows)
})

# 2. Set the visual background style
sns.set_theme(style="whitegrid")

# 3. Create the lmplot with customized marker and regression line colors
plot = sns.lmplot(
    data=df,
    x='price',
    y='reviews',
    height=6,
    aspect=1.3,
    scatter_kws={'color': '#1f77b4', 'alpha': 0.8}, # Customize marker color and transparency
    line_kws={'color': '#ff7f0e', 'linewidth': 2}   # Customize regression line color
)

# 4. Add customized titles and axis labels
plot.set_axis_labels('Product Price (INR)', 'Number of Customer Reviews', fontsize=12)
plot.fig.suptitle('Flipkart: Product Price vs. Number of Reviews', y=1.03, fontsize=14, fontweight='bold')

# 5. Display the plot
plt.show()
```

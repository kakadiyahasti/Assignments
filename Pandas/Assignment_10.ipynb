Create two DataFrames: one called users with columns user_id, username, and city (add at least 5 rows), and another called orders with columns order_id, user_id, and order_amount (add at least 7 rows). Merge them using pd.merge() to show only users who have placed an order (inner join)

```python
users = pd.DataFrame({
    "user_id": [1, 2, 3, 4, 5],
    "username": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "city": ["New York", "London", "Paris", "Tokyo", "Sydney"]
})

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "user_id": [1, 2, 1, 4, 5, 2, 4],
    "order_amount": [250, 120, 75, 300, 180, 90, 450]
})

result = pd.merge(users,orders,on='user_id',how = 'inner')
print(result)
```

Using the same users and orders DataFrames, perform a left join to display all users along with their orders. If a user has not placed any orders, show NaN for order_id and order_amount.<br><br><em><strong>Hint:</strong> Use the how='left' parameter in pd.merge().</em>

```python
users = pd.DataFrame({
    "user_id": [1, 2, 3, 4, 5],
    "username": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "city": ["New York", "London", "Paris", "Tokyo", "Sydney"]
})

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "user_id": [1, 2, 1, 4, 5, 2, 4],
    "order_amount": [250, 120, 75, 300, 180, 90, 450]
})

result = pd.merge(users,orders,on='user_id',how='left')
print(result)
```

Imagine you have two DataFrames: restaurants (restaurant_id, name, city) and zomato_reviews (review_id, restaurant_id, rating, reviewer_city). Merge them using an outer join on restaurant_id to show all restaurants and all reviews, even if some restaurants have no reviews or some reviews are for restaurants not in your list.

```python
restaurants = pd.DataFrame({
    "restaurant_id": [1, 2, 3, 4],
    "name": ["Spice Hub", "Pizza Palace", "Burger House", "Café Delight"],
    "city": ["Delhi", "Mumbai", "Pune", "Bangalore"]
})

zomato_reviews = pd.DataFrame({
    "review_id": [101, 102, 103, 104],
    "restaurant_id": [1, 2, 5, 2],
    "rating": [4.5, 3.8, 4.2, 4.0],
    "reviewer_city": ["Delhi", "Mumbai", "Chennai", "Pune"]
})

result = pd.merge(restaurants,zomato_reviews,on='restaurant_id',how='outer')
print(result)
```

Create two DataFrames: playlists (playlist_id, user_id, genre) and user_profiles (user_id, username, city). Merge them using pd.merge() on both user_id and city to get playlists where the playlist owner's city matches the user profile city.

```python
playlists = pd.DataFrame({
    "playlist_id": [101, 102, 103, 104, 105],
    "user_id": [1, 2, 3, 4, 5],
    "genre": ["Pop", "Rock", "Jazz", "Hip-Hop", "Classical"],
    "city": ["Delhi", "Mumbai", "Pune", "Delhi", "Bangalore"]
})

user_profiles = pd.DataFrame({
    "user_id": [1, 2, 3, 4, 5],
    "username": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "city": ["Delhi", "Mumbai", "Delhi", "Delhi", "Chennai"]
}) 

result = pd.merge(playlists,user_profiles,on='user_id',how='inner')
print(result)
```

Use ChatGPT or Copilot to generate sample code that merges two DataFrames: one for Flipkart products (product_id, name, category) and one for product ratings (rating_id, product_id, rating). Copy the code, run it, and modify it so that it performs a right join instead of the default.

```python
import pandas as pd

# Flipkart products DataFrame
products = pd.DataFrame({
    "product_id": [1, 2, 3, 4],
    "name": ["Laptop", "Smartphone", "Headphones", "Smartwatch"],
    "category": ["Electronics", "Electronics", "Audio", "Wearables"]
})

# Product ratings DataFrame
ratings = pd.DataFrame({
    "rating_id": [101, 102, 103, 104, 105],
    "product_id": [1, 2, 2, 3, 5],
    "rating": [4.5, 4.0, 3.5, 5.0, 4.2]
})

# Right join: keep ALL ratings, even if the product is not in products
merged_df = pd.merge(
    products,
    ratings,
    on="product_id",
    how="right"
)

print(merged_df)
```

Download a publicly available CSV dataset of IPL cricket matches (for example, from Kaggle or data.world), and use pandas to load it into a DataFrame named ipl_df. Display the first 5 rows.

```python
ipl_df = pd.read_csv('C:\\Users\\lenovo\\Desktop\\Assignments\\Pandas\\matches.csv')
ipl_df.head()
```

Clean the ipl_df DataFrame by removing any rows where the 'winner' column is missing or null, and reset the index afterwards.

```python

# Check missing values in winner column
print("Missing values before cleaning:")
print(ipl_df['winner'].isnull().sum())

# Remove rows where winner is missing
ipl_df = ipl_df.dropna(subset=['winner'])

# Reset index
ipl_df = ipl_df.reset_index(drop=True)

# Display first 5 rows after cleaning
print("\nFirst 5 rows after cleaning:")
print(ipl_df.head())

# Check missing values after cleaning
print("\nMissing values after cleaning:")
print(ipl_df['winner'].isnull().sum())

# Display updated dataset shape
print("\nUpdated Dataset Shape:")
print(ipl_df.shape)
```

Transform the ipl_df DataFrame to add a new column 'match_margin_type' that labels each match as 'Runs' if the 'win_by_runs' value is greater than 0, 'Wickets' if 'win_by_wickets' is greater than 0, or 'Tie/No Result' otherwise.<br><br><em><strong>Hint:</strong> Use numpy.select() or pandas' apply() for this transformation.</em>

```python

import numpy as np
import pandas as pd

# Define conditions
conditions = [
    ipl_df['win_by_runs'] > 0,
    ipl_df['win_by_wickets'] > 0
]

# Define corresponding labels
choices = ['Runs', 'Wickets']

# Create new column
ipl_df['match_margin_type'] = np.select(
    conditions,
    choices,
    default='Tie/No Result'
)

# Display selected columns
print(ipl_df[
    ['winner', 'win_by_runs', 'win_by_wickets',
     'match_margin_type']
].head(10))
```

Merge the ipl_df DataFrame with a separate DataFrame containing IPL team names and their home cities (create this manually with at least 5 teams), using the team name as the key, and add the home city for both 'team1' and 'team2' in each match.

```python

import pandas as pd

# Create a DataFrame containing IPL teams and home cities
team_data = {
    'team': [
        'Mumbai Indians',
        'Chennai Super Kings',
        'Royal Challengers Bangalore',
        'Kolkata Knight Riders',
        'Rajasthan Royals',
        'Delhi Daredevils',
        'Kings XI Punjab',
        'Sunrisers Hyderabad'
    ],
    'home_city': [
        'Mumbai',
        'Chennai',
        'Bangalore',
        'Kolkata',
        'Jaipur',
        'Delhi',
        'Mohali',
        'Hyderabad'
    ]
}

teams_df = pd.DataFrame(team_data)

# Merge home city for team1
ipl_df = ipl_df.merge(
    teams_df,
    left_on='team1',
    right_on='team',
    how='left'
)

# Rename home_city column
ipl_df.rename(
    columns={'home_city': 'team1_home_city'},
    inplace=True
)

# Remove extra team column
ipl_df.drop(columns=['team'], inplace=True)

# Merge home city for team2
ipl_df = ipl_df.merge(
    teams_df,
    left_on='team2',
    right_on='team',
    how='left'
)

# Rename home_city column
ipl_df.rename(
    columns={'home_city': 'team2_home_city'},
    inplace=True
)

# Remove extra team column
ipl_df.drop(columns=['team'], inplace=True)

# Display first 10 rows
print("IPL Dataset with Home Cities:")
print(
    ipl_df[
        ['team1', 'team1_home_city',
         'team2', 'team2_home_city']
    ].head(10)
)
```

Create a pivot table to show the total number of matches won by each team per season, and display the top 3 teams with the most wins in any single season

```python

import pandas as pd

# Create pivot table showing wins by each team per season
wins_pivot = pd.pivot_table(
    ipl_df,
    index='winner',
    columns='season',
    values='id',
    aggfunc='count',
    fill_value=0
)

# Display pivot table
print("Total Matches Won by Each Team per Season:")
print(wins_pivot)

# Find top 3 team-season combinations
top_3_wins = (
    wins_pivot
    .stack()
    .reset_index()
)

top_3_wins.columns = ['Team', 'Season', 'Wins']

# Sort by wins in descending order
top_3_wins = top_3_wins.sort_values(
    by='Wins',
    ascending=False
)

# Display top 3 results
print("\nTop 3 Teams with Most Wins in a Single Season:")
print(top_3_wins.head(3))
```

Given a DataFrame of food items with columns: Item, Price, and Qty, use the apply() function with a lambda to create a new column 'DiscountedPrice' that applies a 10% discount to the Price for each item.

```python
food = {
    'Item': ['Apple', 'Banana', 'Milk', 'Bread', 'Eggs'],
    'Price': [120, 40, 60, 45, 90],
    'Qty': [5, 12, 2, 3, 10]
}

df = pd.DataFrame(food)

df['DiscountedPrice'] = df['Price'].apply(lambda x: x * 0.90)
df
```

Given a DataFrame of Zomato restaurant orders with columns: Restaurant, Price, Qty, and DeliveryCharge, use a lambda function with apply() to add a column 'FinalAmount' that calculates the total amount for each order as (Price * Qty) + DeliveryCharge.

```python
data = {
    'Restaurant': ['Dominos', 'Biryani By Kilo', 'Subway', 'Haldiram', 'KFC'],
    'Price': [350, 450, 220, 180, 500],
    'Qty': [2, 1, 3, 2, 1],
    'DeliveryCharge': [40, 50, 30, 25, 45]
}

df = pd.DataFrame(data)

df['FinalAmount'] = df.apply(lambda x : (x['Price'] * x['Qty'] )+ x['DeliveryCharge'], axis = 1)
df
```

Use ChatGPT to generate a pandas code snippet that adds a column 'GST' to a DataFrame of Myntra orders, where GST is calculated as 5% of (Price * Qty). Paste the AI-generated code, run it, and write 2 lines about what you learned from the AI's explanation.

```python

# Sample Myntra order dataset
data = {
    'Order_ID': [101, 102, 103, 104],
    'Item': ['T-Shirt', 'Jeans', 'Jacket', 'Sneakers'],
    'Price': [799, 1999, 2999, 2499],
    'Qty': [2, 1, 1, 2]
}

df = pd.DataFrame(data)

# Calculate GST as 5% of (Price * Qty) using apply() with lambda
df['GST'] = df.apply(lambda row: 0.05 * (row['Price'] * row['Qty']), axis=1)

print(df)
```

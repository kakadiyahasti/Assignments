Perform univariate analysis on the 'runs' column: calculate and print the mean, median, mode, minimum, and maximum runs scored by batsmen in the IPL_2023_Batsmen dataset.

```python
# Univariate analysis of Runs column

# Calculate mean
mean_runs = ipl_df['Runs'].mean()

# Calculate median
median_runs = ipl_df['Runs'].median()

# Calculate mode
mode_runs = ipl_df['Runs'].mode()

# Calculate minimum
min_runs = ipl_df['Runs'].min()

# Calculate maximum
max_runs = ipl_df['Runs'].max()

# Print results
print("Univariate Analysis of IPL 2023 Batsmen - Runs")
print("---------------------------------------------")

print("Mean Runs:", mean_runs)
print("Median Runs:", median_runs)
print("Mode Runs:", mode_runs.tolist())
print("Minimum Runs:", min_runs)
print("Maximum Runs:", max_runs)
```

Create a Pandas DataFrame for your top 5 favorite Zomato restaurants with columns: 'name', 'rating', 'votes', and 'avg_cost'. Perform bivariate analysis between 'rating' and 'votes' by calculating their correlation coefficient.

```python
data = {
    'name': [
        'The Belgian Waffle Co',
        'Dominos Pizza',
        'McDonalds',
        'Subway',
        'Pizza Hut'
    ],
    'rating': [4.2, 4.0, 4.1, 3.9, 4.3],
    'votes': [850, 1200, 1500, 650, 950],
    'avg_cost': [300, 500, 400, 350, 600]
}

# Create DataFrame
zomato_df = pd.DataFrame(data)

# Display DataFrame
print("Zomato Restaurant Data:")
print(zomato_df)

# Bivariate analysis: Rating vs Votes
correlation = zomato_df['rating'].corr(
    zomato_df['votes']
)

# Print correlation coefficient
print("\nCorrelation between Rating and Votes:")
print(correlation)
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set visualization style
sns.set_theme(style="whitegrid")

# Calculate correlation
correlation = ipl_df['Runs'].corr(ipl_df['SR'])

print("Correlation between Runs and Strike Rate:")
print(round(correlation, 3))

# Create scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=ipl_df,
    x='Runs',
    y='SR',
    hue='Team',
    s=100,
    alpha=0.8
)

# Add title and labels
plt.title(
    'Relationship Between Runs and Strike Rate',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Total Runs Scored', fontsize=12)
plt.ylabel('Strike Rate', fontsize=12)

plt.legend(
    title='Team',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()

```

Given a Flipkart-style product reviews dataset with some missing 'rating' values, use pandas to fill all missing ratings with the mean rating of the dataset and print the updated DataFrame.<br><br><em><strong>Hint:</strong> Use the fillna() method with the mean() function.</em>

```python
reviews = pd.DataFrame({
    "product_id": ["PROD002", "PROD003", "PROD004", "PROD005", "PROD006"],
    "product_name": [
        "Smartwatch","Bluetooth Speaker","Gaming Mouse","Mechanical Keyboard","Power Bank",
    ],
    "review_text": [
        "Battery life is weak","Super loud and clear","Very comfortable","Solid build quality","Fast charging works well",
    ],
    "rating": [4.5, np.nan, 5.0, np.nan, 3.5],
})

mean_rating = np.mean(reviews['rating'])
reviews['rating'] = reviews['rating'].fillna(mean_rating)
reviews
```

Use ChatGPT to generate pandas code that replaces all missing values in a 'followers' column (from a simulated Instagram user data CSV) with 0, then test the code on your own sample data and submit both the code and the result.<br><br><em><strong>Hint:</strong> Ask ChatGPT: 'How do I replace missing values in a pandas DataFrame column with zero?'</em>

```python
from io import StringIO

# Simulated Instagram user data CSV
csv_data = """username,followers
alice,1200
bob,
charlie,850
diana,
eve,2300
"""

# Load the CSV into a DataFrame
df = pd.read_csv(StringIO(csv_data))

# Replace missing values in the 'followers' column with 0
df["followers"] = df["followers"].fillna(0)

print(df)
```

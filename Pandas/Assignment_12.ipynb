Given a DataFrame of Zomato food orders with columns ['city', 'restaurant', 'order_amount'], use groupby to calculate the average order_amount for each restaurant in each city using the agg() function.

```python
zomato_orders = pd.DataFrame({
    'city': [
        'Mumbai', 'Mumbai', 'Mumbai', 'Mumbai', 'Mumbai',
        'Delhi', 'Delhi', 'Delhi', 'Delhi', 'Delhi',
        'Bangalore', 'Bangalore', 'Bangalore', 'Bangalore', 'Bangalore',
        'Pune', 'Pune', 'Pune', 'Pune', 'Pune',
        'Ahmedabad', 'Ahmedabad', 'Ahmedabad', 'Ahmedabad', 'Ahmedabad'
    ],
    
    'restaurant': [
        "Domino's", "KFC", "Domino's", "McDonald's", "KFC",
        "Domino's", "KFC", "McDonald's", "Domino's", "KFC",
        "McDonald's", "KFC", "Domino's", "McDonald's", "KFC",
        "Domino's", "McDonald's", "KFC", "Domino's", "KFC",
        "McDonald's", "Domino's", "KFC", "McDonald's", "Domino's"
    ],
    
    'order_amount': [
        450, 380, 520, 600, 420,
        350, 450, 550, 400, 500,
        700, 480, 550, 650, 520,
        300, 450, 400, 350, 500,
        600, 400, 450, 550, 500
    ]
})

avg_orders = zomato_orders.groupby(['city', 'restaurant']).agg(avg_order_amount=('order_amount', 'mean'))
print(avg_orders)
```

Take a Flipkart-style product sales DataFrame with columns ['category', 'region', 'sales', 'units_sold'], and apply multiple aggregations: for each (category, region) pair, calculate the total sales and the average units_sold using groupby and agg().

```python
sales = pd.DataFrame ({
    'category': ['Electronics', 'Electronics', 'Electronics', 'Fashion', 'Fashion', 'Fashion', 'Electronics', 'Fashion'],
    'region': ['North', 'North', 'South', 'North', 'South', 'South', 'South', 'North'],
    'sales': [45000, 32000, 55000, 1200, 2500, 1800, 48000, 3100],
    'units_sold': [15, 10, 20, 4, 8, 5, 16, 9]
})

result = sales.groupby(['category','region']).agg(total_sales=('sales','sum'),avg_unit_sold=('units_sold','mean'))
result
```

Refactor the code that groups a DataFrame by 'region' and sums 'revenue' so it also returns the maximum and minimum revenue per region in the same result.<br><br><em><strong>Hint:</strong> Use agg() with a dictionary to specify multiple aggregation functions for the 'revenue' column.</em>

```python
result = sales.groupby('region').agg(
    {'sales':['sum','min','max']}
)
result
```

Suppose you have a DataFrame of Spotify song streams with columns ['artist', 'genre', 'streams']. Use groupby and agg() to find both the total and average streams for each genre, then sort the result by total streams in descending order.<br><br><em><strong>Constraint:</strong> Your final DataFrame should have genres as the index and columns for total and average streams.</em>

```python
songs = pd.DataFrame({
    'artist': ['Taylor Swift', 'The Weeknd', 'Bad Bunny', 'Drake', 'Ed Sheeran', 
               'Dua Lipa', 'BTS', 'Olivia Rodrigo', 'Billie Eilish', 'Post Malone'],
    'genre': ['Pop', 'R&B', 'Reggaeton', 'Hip-Hop', 'Pop', 
              'Pop', 'K-Pop', 'Pop', 'Pop', 'Hip-Hop'],
    'streams': [45000000, 38000000, 52000000, 49000000, 29000000, 
                31000000, 41000000, 27000000, 33000000, 36000000]
})

result = songs.groupby('genre').agg(
    total_streams = ('streams','sum'),
    avg_streams = ('streams','mean')
).sort_values('total_streams',ascending = False)

result
```

Given a pandas DataFrame of Flipkart product reviews with columns: product_id, user_id, rating, review_text, identify and print the number of duplicate reviews (where product_id and user_id are both the same) before and after using drop_duplicates().

```python
reviews = {
    'product_id': ['FKP001', 'FKP002', 'FKP001', 'FKP003', 'FKP002', 'FKP004', 'FKP001', 'FKP005', 'FKP003', 'FKP004'],
    'user_id':    ['USR101', 'USR102', 'USR101', 'USR103', 'USR102', 'USR104', 'USR105', 'USR106', 'USR103', 'USR107'],
    'rating':     [5,        4,        5,        3,        2,        5,        4,        1,        3,        5],
    'review_text': [
        'Great phone, highly recommended!',
        'Good quality product.',
        'Great phone, highly recommended!',  
        'Average build quality.',
        'Faced issues with display.',       
        'Value for money.',
        'Decent battery life.',
        'Defective product received.',
        'Average build quality.',           
        'Superb performance!'
    ]
}

df = pd.DataFrame(reviews)
df

duplicated_no = df.duplicated(subset=['product_id','user_id']).sum()

df= df.drop_duplicates(subset=['product_id','user_id'],keep='first')

duplicate_after_drop = df.duplicated(subset=['product_id','user_id']).sum()

print("duplicate count before drop: ",duplicated_no)
print("duplicate count after drop: ",duplicate_after_drop)
```

Use the IQR method to detect outliers in the 'order_amount' column of a Zomato orders DataFrame and print the order IDs that are considered outliers.<br><br><em><strong>Hint:</strong> Calculate Q1, Q3, and IQR, then filter orders outside the [Q1 - 1.5*IQR, Q3 + 1.5*IQR] range.</em>

```python
sample_data = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "order_amount": [100, 120, 95, 110, 105, 115, 1000]
})

q1 = np.quantile(sample_data['order_amount'],0.25)

q3 = np.quantile(sample_data['order_amount'],0.75)
iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print(lower_bound)
print(upper_bound)
```

Use ChatGPT or Copilot to generate a pandas function that finds and returns all outlier transaction amounts in a Paytm wallet DataFrame using the Z-score method, then test the function with sample data.

```python
def find_outlier_transactions(df, column="transaction_amount", threshold=1.5):
    
    amounts = df[column]

    mean = amounts.mean()
    std = amounts.std()

    z_scores = (amounts - mean) / std

    return df[abs(z_scores) > threshold]

sample_data = pd.DataFrame({
    "transaction_id": [101, 102, 103, 104, 105, 106, 107],
    "transaction_amount": [100, 120, 95, 110, 105, 115, 1000]
})

outliers = find_outlier_transactions(
    sample_data,
    column="transaction_amount",
    threshold=1.5
)

print(outliers)
```

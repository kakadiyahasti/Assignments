Use pd.concat to combine two DataFrames representing January and February orders from a Swiggy-style food delivery app, stacking them row-wise to create a single DataFrame with all orders.

```python
january_orders = pd.DataFrame({
    'order_id': [1001, 1002, 1003],
    'restaurant': ['Burger King', 'Pizza Hut', 'Subway'],
    'amount': [350, 620, 240]
})

# Sample February orders dataset
february_orders = pd.DataFrame({
    'order_id': [1004, 1005, 1006],
    'restaurant': ['KFC', 'Dominos', 'Taco Bell'],
    'amount': [480, 550, 310]
})

result = pd.concat([january_orders,february_orders], ignore_index=True)
result
```

Given two DataFrames — one with Flipkart product details and another with their respective ratings — use pd.concat to merge them column-wise so each product row includes its rating.

```python
df_products = pd.DataFrame({
    'product_id': ['F101', 'F102', 'F103', 'F104'],
    'product_name': ['Wireless Earbuds', 'Smartwatch', 'Gaming Mouse', 'Bluetooth Speaker'],
    'price': [1999, 2999, 899, 1499]
})

# Sample ratings DataFrame
df_ratings = pd.DataFrame({
    'rating': [4.5, 4.2, 4.8, 3.9],
    'reviews_count': [1250, 840, 3100, 450]
})

result = pd.concat([df_products,df_ratings],axis=1)
result
```

Create a DataFrame showing daily song streams for three users on Spotify for one week, then use the pivot function to reshape the data so each row is a user and each column is a day.

```python
data = {
    'user': ['Alice', 'Alice', 'Alice', 'Alice', 'Alice', 'Alice', 'Alice',
             'Bob', 'Bob', 'Bob', 'Bob', 'Bob', 'Bob', 'Bob',
             'Charlie', 'Charlie', 'Charlie', 'Charlie', 'Charlie', 'Charlie', 'Charlie'],
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'] * 3,
    'streams': [12, 15, 8, 20, 25, 30, 18,
                5, 10, 14, 12, 18, 22, 15,
                20, 22, 19, 25, 35, 40, 28]
}

df = pd.DataFrame(data)

result = pd.pivot(data=df,index='day', columns='user', values='streams')
result
```
